### Q1 - Fare Analysis with Array Iteration

Use the Fare column and create a NumPy array.

Perform the following using array iteration:
1. Print every Fare value.  
2. Print only Fares greater than ₹100.  
3. Count passengers whose Fare is greater than ₹100.  
4. Calculate the total Fare using iteration.

```python
fare = df['Fare'].to_numpy()
```

```python
#1
for value in fare:
  print(value)
```

```python
#2
for value in fare:
  if value > 100:
    print(value)
```

```python
#3
count = 0

for value in fare:
  if value > 100:
    count += 1

print(count)
```

```python
#4
total = 0

for value in fare:
  total += value

print(total)
```

### Q8 — Copy vs View
Create a NumPy array using the first 10 Fare values.

Create:

 one view  
 one copy  

Modify the first value of the view by adding ₹100.
Then check whether the original array changed.

After that, modify the first value of the copy by adding ₹200 and check whether the
original array changed.

```python
fare = df['Fare'][:10].to_numpy()

fare
```

```python
view = fare.view()
copy = fare.copy()

print(view)
print(copy)
```

```python
view[0] = 100

view
```

### Q9 — Broadcasting Fare Adjustment
Create a matrix containing:
Pclass, Age, SibSp, Parch, Fare

Increase only the Fare column by 10% using broadcasting.

Then:
1. Display the first 10 rows before modification.  
2. Display the first 10 rows after modification.  
3. Find the new highest Fare.

```python
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()
```

```python
mat = np.column_stack((pclass, age, sibsp, parch, fare))

mat
```

```python
print(mat[:10])
```

```python
mat[:, 4] = mat[:, 4] * 10 / 100 + mat[:, 4]

mat
```

```python
print(mat[:10])
```

```python
print(np.max(mat[:, 4]))
```
