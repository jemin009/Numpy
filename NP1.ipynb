```python
#7
"""
Create a NumPy array and find: 
 Total revenue  
 Average monthly revenue 
"""

revenue = [45000, 52000, 48000, 61000, 58000, 70000] 

total = np.array(revenue).sum()
avg = np.array(revenue).mean()

print(f"Total revenue : {total}")
print(f"Average monthly revenue : {avg}")
```

```python
#8
"""
Convert it into a NumPy array. 
Then try multiplying the NumPy array by 2 and observe the result. 
Also try multiplying the original Python list by 2. 
Explain the difference in the output.
"""

data = [10, 20, 30, 40, 50] 

arr = np.array(data)

arr_multi = arr * 2  # every element is multiply by 2
list_multi = data * 2  # whole list multiply by 2

print(arr_multi) 
print(list_multi)
```

```python
#9 
"""
You have the following number of orders received by a company each day: 

Using NumPy: 
1. Convert the data into an array.  
2. Find total orders.  
3. Find average daily orders. 
"""

orders = [120, 135, 98, 150, 175, 160, 142] 

total = np.array(orders).sum()
avg = np.array(orders).mean()

print(f"Total orders : {total}")
print(f"Average daily orders : {avg}")
```

```python
#10
"""
Using NumPy, create an array and find: 
1. Total sales  
2. Average sales  
3. Highest sales  
4. Lowest sales
"""

sales = [250, 400, 350, 500, 450] 

total = np.array(sales).sum()
avg = np.array(sales).mean()
highest = np.array(sales).max()
lowest = np.array(sales).min()

print(f"Total sales : {total}")
print(f"Average sales : {avg}")
print(f"Highest sales : {highest}")
print(f"Lowest sales : {lowest}")
```
