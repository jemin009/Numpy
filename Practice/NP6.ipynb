```python
import numpy as np
```

```python
#1
"""
Perform: 
1. Create the NumPy array.  
2. Find total visitors.  
3. Find average visitors per hour.  
4. Find the busiest hour's visitor count.  
5. Extract visitors from hour 4 to hour 9.  
6. Extract the last 4 hours.  
7. Increase every value by 10% to simulate projected traffic.  
8. Find the maximum projected traffic.  
9. Check whether projected traffic crosses 330.  
10. Print shape and ndim. 
"""

visitors = [120, 145, 132, 178, 210, 195, 230, 260, 245, 280, 310, 295] 

arr = np.array(visitors)

increased = arr + (arr * 10) / 100

print(f"total visitors : {arr.sum()}")
print(f"average visitors : {arr.mean()}")
print(f"the busiest hour's visitor count : {arr.max()}")
print(f"visitors from hour 4 to hour 9 : {arr[3:9]}")
print(f"the last 4 hours : {arr[-4:]}")
print(f"maximum projected traffic : {increased.max()}")

if (increased > 330).any():
    print("Yes, projected traffic crosses 330")
else:
    print("No, projected traffic doesn't crosses 330")

print(f"Shape : {increased.shape}")
print(f"Ndim : {increased.ndim}")
```

```python
#2
"""
Find: 
1. Average delivery time.  
2. Fastest delivery.  
3. Slowest delivery.  
4. Extract the first 5 orders.  
5. Extract the last 3 orders.  
6. Extract orders 3–7.  
7. Add 5 minutes to every delivery time as a simulated traffic delay.  
8. Find the new average.  
9. Find the new maximum.  
10. Check whether the delayed maximum exceeds 60 minutes. 
"""

delivery_time = [32, 45, 28, 51, 39, 42, 55, 31, 48, 36] 

arr = np.array(delivery_time)

new = arr + 5

print(f"Average delivery time : {arr.mean()}")
print(f"Fastest delivery time : {arr.min()}")
print(f"Slowest delivery time : {arr.max()}")
print(f"the first 5 orders : {arr[:5]}")
print(f"the last 3 orders : {arr[-3:]}")
print(f"orders 3–7 : {arr[2:7]}")
print(f"new average. : {new.mean()}")
print(f"new maximum. : {new.max()}")

if (new > 60).any():
    print("Yes, the delayed maximum exceeds 60 minutes")
else:
    print("No, the delayed maximum not exceeds 60 minutes")
```

```python
#3
"""
Tasks: 
1. Create the 2D array.  
2. Find total annual expense of each department.  
3. Find the overall average expense.  
4. Find the maximum expense.  
5. Find the minimum expense.  
6. Extract Feb and Mar expenses for all departments.  
7. Extract all months of Department 2.  
8. Find the total expense of each department using axis=1.  
9. Check whether any monthly expense exceeds ₹50,000. 

Rows = Departments 
Columns = Jan, Feb, Mar, Apr 
"""

expenses = [ 
[45000, 47000, 49000, 51000], 
[32000, 35000, 34000, 38000], 
[28000, 30000, 31000, 33000] 
] 

arr = np.array(expenses)

print(f"total annual expense of each department : {arr.sum(axis=1)}")
print(f"the overall average expense : {arr.mean()}")
print(f"the maximum expense : {arr.max()}")
print(f"the minimum expense : {arr.min()}")
print(f"Feb and Mar expenses for all departments : {arr[:,1:3]}")
print(f"all months of Department 2 : {arr[1]}")
print(f"total expense of each department using axis=1 : {arr.sum(axis=1)}")

if (arr > 50000).any():
    print("Yes, monthly expense exceeds ₹50,000")
else:
    print("No, monthly expense not exceeds ₹50,000")
```

```python
#4
"""
Find: 
1. Create both arrays.  
2. Average conversions for A.  
3. Average conversions for B.  
4. Maximum conversion for A.  
5. Maximum conversion for B.  
6. Combine the two arrays into a 2D array.  
7. Print its shape and ndim.  
8. Find the total conversions for each version using axis=1.  
9. Increase Version B conversions by 10%.  
10. Find the new average of Version B. 
"""

A = [42, 48, 51, 46, 55, 60, 58] 
B = [45, 52, 49, 57, 61, 64, 68] 

arr_A = np.array(A)
arr_B = np.array(B)

print(f"Average conversions for A : {arr_A.mean()}")
print(f"Average conversions for B : {arr_B.mean()}")

print(f"Maximum conversions for A : {arr_A.max()}")
print(f"Maximum conversions for B : {arr_B.max()}")

combine = np.array([A,B])
print(f"Combine array : {combine}")

print(f"Shape : {combine.shape}")
print(f"Ndim : {combine.ndim}")

print(f"total conversions for each version : {combine.sum(axis=1)}")

increased = arr_B + (arr_B * 10) / 100
print(f"Increased version of B : {increased}")

print(f"new average of Version B : {increased.mean()}")
```

```python
#5
"""
Find: 
1. Create the array.  
2. Add 20 units to every product.  
3. Find the new inventory.  
4. Find average inventory.  
5. Find maximum and minimum inventory.  
6. Extract products 2–6.  
7. Extract the last 3 products.  
8. Check whether the minimum inventory is below 30.  
9. Print shape and ndim.  

A store has inventory for 8 products: 
The store receives 20 additional units for every product. 
"""

inventory = [45, 12, 78, 23, 8, 56, 15, 92] 

arr = np.array(inventory)

add = arr + 20

print(f"new inventory : {add}")
print(f"average inventory : {add.mean()}")
print(f"maximum inventory : {add.max()}")
print(f"minimum inventory : {add.min()}")
print(f"products 2–6 : {add[1:6]}")
print(f"last 3 products : {add[-3:]}")

if add.min() < 30:
    print("Yes, the minimum inventory is below 30")
else:
    print("No, the minimum inventory is not below 30")

print(f"Shape : {add.shape}")
print(f"Ndim : {add.ndim}")
```

```python
#6
"""
Find: 
1. Shape and ndim.  
2. Marks of Student 4 in Python.  
3. All subjects of Student 2.  
4. Maths and Statistics marks of all students.  
5. Total marks of every student using axis=1.  
6. Average marks of every student using axis=1.  
7. Highest individual mark.  
8. Lowest individual mark.  
9. Add 5 marks to every student's Python score only.  
10. Print the updated matrix. 

Rows = Students 
Columns = Maths, Statistics, Python 
"""

marks = [ 
[72, 68, 81], 
[85, 78, 90], 
[64, 71, 69], 
[91, 88, 94], 
[76, 82, 79] 
]

arr = np.array(marks)

print(f"Shape : {arr.shape}")
print(f"Ndim : {arr.ndim}")
print(f"Marks of Student 4 in Python : {arr[3][-1]}")
print(f"All subjects of Student 2 : {arr[1]}")
print(f"Maths and Statistics marks of all students : {arr[:,:2]}")
print(f"Total marks of every student : {arr.sum(axis=1)}")
print(f"Average marks of every student : {arr.mean(axis=1)}")
print(f"Highest individual mark : {arr.max(axis=1)}")
print(f"Lowest individual mark : {arr.min(axis=1)}")

arr[:,-1] = arr[:,-1] + 5
print(f"updated matrix : {arr}")
```

```python
#7
"""
Find: 
1. Shape and ndim.  
2. Store 2 → Product 1 → Quarter 3.  
3. All quarters of Store 1 → Product 2.  
4. All products of Store 3 → Quarter 4.  
5. Maximum sales value.  
6. Minimum sales value.  
7. Extract all products and quarters of Store 2.  
8. Find total sales of Store 1.  
9. Check whether maximum sales exceed ₹24,000.  

Dimensions: 
Store → Product → Quarter 
"""

sales = [ 
    [ 
        [12000, 15000, 18000, 21000], 
        [10000, 13000, 16000, 19000] 
    ], 
    [ 
        [14000, 17000, 20000, 23000], 
        [11000, 15000, 18000, 22000] 
    ], 
    [ 
        [16000, 19000, 22000, 25000], 
        [13000, 16000, 21000, 24000] 
    ] 
]

arr = np.array(sales)

print(f"Shape : {arr.shape}")
print(f"Ndim : {arr.ndim}")
print(f"Store 2 → Product 1 → Quarter 3 : {arr[1][0][2]}")
print(f"All quarters of Store 1 → Product 2 : {arr[0][1]}")
print(f"All products of Store 3 → Quarter 4 : {arr[2][:,-1]}")
print(f"Maximum sales value : {arr.max()}")
print(f"Minimum sales value : {arr.min()}")
print(f"all products and quarters of Store 2 : {arr[1]}")
print(f"total sales of Store 1 : {arr[0].sum()}")

if arr.max() > 24000:
    print("Yes, maximum sales exceed ₹24,000")
else:
    print("No, maximum sales not exceed ₹24,000")
```
