```python
#1
"""
Using NumPy: 
1. Create a NumPy array.  
2. Find total sales.  
3. Find average sales.  
4. Find highest sales.  
5. Find lowest sales.  
6. Find the last day's sales using negative indexing.  
7. Extract the first 3 days using slicing.  
8. Print shape, ndim, and dtype.  
"""

import numpy as np

sales = [12500, 14800, 13200, 17600, 15900, 19200, 18500] 

arr = np.array(sales)

print(arr)
print(f"Total sales : {arr.sum()}")
print(f"Average sales : {arr.mean()}")
print(f"Highest sales : {arr.max()}")
print(f"Lowest sales : {arr.min()}")
print(f"Last day's sales : {arr[-1]}")
print(f"First three days : {arr[:3]}")
print(f"Shape : {arr.shape}")
print(f"Ndim : {arr.ndim}")
print(f"Dtype : {arr.dtype}")
```

```python
#2
"""
Using NumPy: 
1. Create the array.  
2. Find average rating.  
3. Find highest and lowest rating.  
4. Extract the last 4 ratings using slicing.  
5. Extract the second rating using indexing.  
6. Check the dtype and ndim.  
7. Determine whether the average rating is ≥ 4.0.
"""

ratings = [4.2, 3.8, 4.5, 4.9, 3.7, 4.6, 4.1, 4.8] 

arr = np.array(ratings)

print(f"Average rating : {arr.mean()}")
print(f"Highest rating : {arr.max()}")
print(f"Lowest rating : {arr.min()}")
print(f"Last 4 ratings : {arr[-4:]}")
print(f"Second ratings : {arr[1]}")
print(f"dtype : {arr.dtype}")
print(f"ndim : {arr.ndim}")

if arr.mean() >= 4.0:
    print(f"Yes, the average rating is >= 4.0")
else:
    print(f"No, the average rating is not >= 4.0")
```

```python
#5
"""
Using NumPy: 
1. Create the array.  
2. Increase every price by ₹100.  
3. Find the new average price.  
4. Find the highest new price.  
5. Extract the first 4 new prices.  
6. Find the last price using negative indexing.  
7. Print the datatype of the array.  
"""

prices = [499, 799, 999, 1299, 1499, 1799] 

arr = np.array(prices)

new = arr + 100

print(f"Increased price : {new}")
print(f"Average price : {new.mean()}")
print(f"Highest price : {new.max()}")
print(f"the first 4 new prices : {new[:4]}")
print(f"last price using negative indexing : {new[-1]}")
print(f"the datatype of the array : {new.dtype}")
```

```python
#6
"""
Using NumPy: 
1. Create the array.  
2. Add 2 to every temperature.  
3. Find the average of the updated temperatures.  
4. Find the highest updated temperature.  
5. Extract values from index 2 to index 7.  
6. Find the second-last temperature using negative indexing.  
7. Determine whether the updated maximum temperature is above 37.  
"""

temperature = [28, 30, 31, 29, 32, 35, 33, 36, 34, 30] 

arr = np.array(temperature)

new = arr + 2

print(new)
print(f"Average of the updated temperatures : {new.mean()}")
print(f"highest updated temperature : {new.max()}")
print(f"values from index 2 to index 7 : {new[2:8]}")
print(f"second-last temperature using negative indexing : {new[-2]}")

if new.max() > 37:
    print("Yes, the updated maximum temperature is above 37")
else:
    print("No, the updated maximum temperature is not above 37")
```

```python
#8
"""
Create the following arrays using NumPy: 
1. A 3 × 4 array containing only zeros using zeros().  
2. A 2 × 5 array containing only ones using ones().  
3. A 3 × 3 array where every value is 100 using full().  
4. Create a 2 × 3 array using empty().  
5. Print the shape and ndim of each array.  
"""

print("Zeros :")
print(np.zeros([3,4]))
print(f"Shape : {np.zeros([3,4]).shape}")
print(f"Ndim : {np.zeros([3,4]).ndim}")
print()

print("Ones :")
print(np.ones([2,5]))
print(f"Shape : {np.ones([2,5]).shape}")
print(f"Ndim : {np.ones([2,5]).ndim}")
print()

print("Full :")
print(np.full((3,3),100))
print(f"Shape : {np.full((3,3),100).shape}")
print(f"Ndim : {np.full((3,3),100).ndim}")
print()

print("Empty")
print(np.empty([2,3]))
print(f"Shape : {np.empty([2,3]).shape}")
print(f"Ndim : {np.empty([2,3]).ndim}")
print()
```

```python
#9
"""
Using NumPy: 
1. Create an integer array for sales.  
2. Create a floating-point array for ratings.  
3. Print the dtype of both arrays.  
4. Convert the sales array into float.  
5. Print the new datatype.  
6. Calculate the average of both arrays. 
"""

sales = [100, 200, 300, 400] 
ratings = [4.2, 4.5, 3.8, 4.9]

sales_arr = np.array(sales)
ratings_arr = np.array(ratings)

Converted = sales_arr.astype(float)

print(f"dtype of sales_arr : {sales_arr.dtype}")
print(f"dtype of ratings_arr : {ratings_arr.dtype}")
print(f"Converted : {Converted}")
print(f"New datatypes : {Converted.dtype}")

print(f"Average of sales_arr : {sales_arr.mean()}")
print(f"Average of ratings_arr : {ratings_arr.mean()}")
```

```python
#10
"""
Using NumPy: 
1. Create the 2D array.  
2. Print shape, ndim, and dtype.  
3. Find the total annual sales of Product B.  
4. Find the Q4 sales of Product C using indexing.  
5. Extract Q1 and Q2 sales for all products using slicing.  
6. Find the highest quarterly sales in the entire dataset.  
7. Increase all sales by 10%.  
8. Find the new highest sales value after the increase.  
9. Determine which product has the highest original annual sales.

Rows = Products A, B, C 
Columns = Q1, Q2, Q3, Q4
"""

sales = [ 
    [12000, 15000, 18000, 21000], 
    [14000, 16000, 19000, 23000], 
    [11000, 17000, 20000, 25000] 
]

arr = np.array(sales)

increased = arr + (arr * 10) / 100

print(f"Shape : {arr.shape}")
print(f"Ndim : {arr.ndim}")
print(f"Dtype : {arr.dtype}")
print(f"The total annual sales of product B : {arr[1].sum()}")
print(f"Q4 sales of Product C using indexing : {arr[2][-1]}")
print(f"Q1 and Q2 sales for all products using slicing : {arr[:,:2]}")
print(f"the highest quarterly sales in the entire dataset : {arr.max()}")
print(f"all sales by 10% : {increased}")
print(f"highest sales value after the increase : {increased.max()}")

annual_sales = arr.sum(axis=1)

print(annual_sales)
print(np.argmax(annual_sales))
```
