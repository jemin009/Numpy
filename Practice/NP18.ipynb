```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1 Using Pandas, calculate the average Fare and average Age for each Pclass using groupby() and agg().

avg = df.groupby(['Pclass']).agg(Average_Fare = ('Fare','mean'),Average_Age = ('Age','mean'))

print(avg)
```

```python
#2 Using Pandas string operations, find passengers whose Name contains "Mr." and display Name, Age, Fare, and Survived.

mr = df['Name'].str.contains('Mr.')

show = df.loc[mr,['Name','Age','Fare','Survived']]

print(show)
```

```python
#3 Create a new column called FamilySize using SibSp and Parch, then display the 10 passengers with the smallest FamilySize along with Name, FamilySize, and Survived.

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

show = df.sort_values('FamilySize', ascending=True)
show = show[['Name', 'FamilySize', 'Survived']].head(10)

print(show)
```

```python
#4 Find the average Fare of female passengers who survived and the average Fare of female passengers who did not survive.

female_survived = df.loc[(df['Sex'] == 'female') & (df['Survived'] == 1),'Fare'].mean()

female_not_survived = df.loc[(df['Sex'] == 'female') & (df['Survived'] == 0),'Fare'].mean()

print(female_survived)
print(female_not_survived)
```

```python
#5 Convert Fare to a NumPy array and find the indexes of passengers whose Fare is between 50 and 100, inclusive. Use those indexes to display their Name and Fare.

fare = df['Fare'].to_numpy()

ind = np.where((fare >= 50) & (fare <= 100))[0]

show = df.loc[ind,['Name','Fare']]

print(show)
```

```python
#6 
"""
Using NumPy, create a FarePerPerson value using: 
Fare / FamilySize 

Then display the 10 passengers with the highest FarePerPerson, including Name, 
Fare, FamilySize, and FarePerPerson. 
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

df['FarePerPerson'] = df['Fare'] / df['FamilySize']


show = df.sort_values('FarePerPerson', ascending=False)

show = show[['Name','Fare','FarePerPerson','FamilySize']].head(10)

print(show)
```

```python
#7 
"""
Create a NumPy feature matrix containing: 
Pclass, Age, SibSp, Parch, Fare, FamilySize 

Replace missing Age values with the median Age, then print the matrix shape and 
first 5 rows.
"""

pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()
familysize = df['FamilySize'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

mat = np.column_stack((pclass, age, sibsp, parch, fare, familysize))

print(mat[:5])
print(mat.shape)
```

```python
#8 
"""
Using groupby() and agg(), calculate for each Sex: 
 Minimum Fare  
 Maximum Fare  
 Average Fare  
 Median Age 
"""

grp = df.groupby('Sex').agg(
                            Minfare = ('Fare','min'),
                            Maxfare = ('Fare', 'max'),
                            Avgfare = ('Fare','mean'),
                            Medianage = ('Age','median'))

print(grp)
```

```python
#9 
"""
Create a NumPy Fare array and calculate the percentage difference of every 
passenger's Fare from the average Fare: 
((Fare - average Fare) / average Fare) × 100 

Create a new column called Fare_Percent_Diff and display the 5 passengers with the 
largest absolute percentage difference.
"""

fare = df['Fare'].to_numpy()

df['Fare_Percent_Diff'] = ((fare - df['Fare'].mean()) / df['Fare'].mean()) * 100

index = df['Fare_Percent_Diff'].abs().sort_values(ascending=False).head(5).index

sort = df.loc[index,['Name', 'Fare', 'Fare_Percent_Diff']]

print(sort)
```

```python
#10
"""
Create a final NumPy matrix containing: 
Survived, Pclass, Age, SibSp, Parch, Fare, FamilySize, FarePerPerson 

Then: 
 Replace missing Age with median Age.  
 Create FamilySize.  
 Create FarePerPerson.  
 Min-max normalize Fare.  
 Standardize FarePerPerson.  
 Convert everything into a NumPy matrix.  
 Transpose the matrix.  
 Print the final matrix shape.
"""

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
familysize = df['FamilySize'].to_numpy()

df['FarePerPerson'] = df['Fare'] / df['FamilySize']
fareperperson = df['FarePerPerson'].to_numpy()

fare = (fare - np.min(fare)) / (np.max(fare) - np.min(fare))

age = np.where(np.isnan(age), np.nanmedian(age), age)

fareperperson = (fareperperson - np.mean(fareperperson)) / np.std(fareperperson)

mat = np.column_stack((survived, pclass, age, sibsp, parch, fare, familysize, fareperperson))

trans = mat.T

print(trans.shape)
```
