```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1 Using Pandas, calculate the average Fare and average Age for each Pclass using groupby() and agg().

avg = df.groupby(['Pclass']).agg(Average_Fare = ('Fare','mean'),Average_Age = ('Age','mean'))

print(avg)
```

```python
#2 Using Pandas string operations, find passengers whose Name contains "Mr." and display Name, Age, Fare, and Survived.

mr = df['Name'].str.contains('Mr.')

show = df.loc[mr,['Name','Age','Fare','Survived']]

print(show)
```

```python
#3 Create a new column called FamilySize using SibSp and Parch, then display the 10 passengers with the smallest FamilySize along with Name, FamilySize, and Survived.

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

show = df.sort_values('FamilySize', ascending=True)
show = show[['Name', 'FamilySize', 'Survived']].head(10)

print(show)
```

```python
#4 Find the average Fare of female passengers who survived and the average Fare of female passengers who did not survive.

female_survived = df.loc[(df['Sex'] == 'female') & (df['Survived'] == 1),'Fare'].mean()

female_not_survived = df.loc[(df['Sex'] == 'female') & (df['Survived'] == 0),'Fare'].mean()

print(female_survived)
print(female_not_survived)
```

```python
#5 Convert Fare to a NumPy array and find the indexes of passengers whose Fare is between 50 and 100, inclusive. Use those indexes to display their Name and Fare.

fare = df['Fare'].to_numpy()

ind = np.where((fare >= 50) & (fare <= 100))[0]

show = df.loc[ind,['Name','Fare']]

print(show)
```

```python
#6 
"""
Using NumPy, create a FarePerPerson value using: 
Fare / FamilySize 

Then display the 10 passengers with the highest FarePerPerson, including Name, 
Fare, FamilySize, and FarePerPerson. 
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

df['FarePerPerson'] = df['Fare'] / df['FamilySize']


show = df.sort_values('FarePerPerson', ascending=False)

show = show[['Name','Fare','FarePerPerson','FamilySize']].head(10)

print(show)
```
