```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1 Create a NumPy array from Fare and find the 5 most frequent Fare values along with their frequencies.

fare = df['Fare'].to_numpy()

unique ,counts = np.unique(fare, return_counts=True)

top5 = np.argsort(counts)[-5:][::-1]

print(unique[top5])
print(counts[top5])
```

```python
#2 Create a NumPy array from Age and find the index positions of passengers whose age is between 18 and 25, excluding missing ages.

age = df['Age'].to_numpy()

print(np.where((age >= 18) & (age <= 25)))
```

```python
#3 Create NumPy arrays for Survived and Pclass and calculate the survival rate of Class 1 passengers.

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()

class1_survived = np.sum((survived == 1) & (pclass == 1))
class1_total = np.sum(pclass == 1)

survival_rate = (class1_survived / class1_total) * 100

print(survival_rate)
```

```python
#4 Create NumPy arrays for SibSp and Parch and create a new array representing the total family connections for every passenger.

sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()

fa_con = sibsp + parch

print(fa_con)
```

```python
#5 Create a NumPy array from Fare and find the indices of the 10 passengers who paid the highest fares, ordered from highest to lowest fare. 

fare = df['Fare'].to_numpy()

highest_fare = np.argsort(fare)[-10:][::-1]

print(highest_fare)
```

```python
#6 Create a matrix containing Pclass, Age, SibSp, Parch, and Fare. Replace missing Age values with the median Age and sort the complete matrix by Fare from lowest to highest.

pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()

age = np.where(np.isnan(age), np.median(age), age)

mat = np.column_stack((pclass,age,sibsp,parch,fare))

sorted_mat = mat[np.argsort(mat[:, 4])]

print(sorted_mat)
```

```python
#7 
"""
Create a NumPy array from Fare and apply this rule using vectorized operations: 
 Fare > 200 → increase by 5%  
 Fare ≤ 200 → increase by 10% 
"""

fare = df['Fare'].to_numpy()

new = np.where(fare > 200, fare + (fare * 5) / 100, fare + (fare * 10) / 100)

print(new)
```

```python
#8 Create NumPy arrays for Age and Pclass. Replace missing Age values with the median Age of their respective passenger class.

age = df['Age'].to_numpy()
pclass = df['Pclass'].to_numpy()

median1 = np.nanmedian(age[pclass == 1])
median2 = np.nanmedian(age[pclass == 2])
median3 = np.nanmedian(age[pclass == 3])

age = np.where(np.isnan(age) & (pclass == 1), median1, age)
age = np.where(np.isnan(age) & (pclass == 2), median1, age)
age = np.where(np.isnan(age) & (pclass == 3), median1, age)

print(age)
```

```python
#9 Create a feature matrix containing Pclass, Age, SibSp, Parch, and Fare. Replace missing Age values with the median and standardize both Age and Fare using vectorized operations. 

pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

age = (age - np.nanmean(age)) / np.nanstd(age)
fare = (fare - np.nanmean(fare)) / np.nanstd(fare)

mat = np.column_stack((pclass, age, sibsp, parch, fare))
print(mat)
```

```python
#10 Create a feature matrix containing Survived, Pclass, Age, SibSp, Parch, and Fare. Replace missing Age values with its median, min-max normalize Fare, and then transpose the final matrix. 

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()

fare_min = np.min(fare)
fare_max = np.max(fare)

fare = (fare - fare_min) / (fare_max - fare_min)

age = np.where(np.isnan(age), np.nanmedian(age), age)

mat = np.column_stack((survived, pclass, age, sibsp, parch, fare))

mat = mat.T

print(mat)
```
