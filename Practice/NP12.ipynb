```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1 Create a NumPy array from the Fare column and find the 5 highest fare values using NumPy. 

fare = df['Fare'].to_numpy()

print(np.sort(fare)[-5:])
```

```python
#2 Using the Fare NumPy array, find the indices of the 10 passengers with the highest fares using argsort().

print(np.argsort(fare)[-10:])
```

```python
#3 Create a NumPy array from Pclass and find all unique passenger classes. 

pclass = df['Pclass'].to_numpy()

print(np.unique(pclass))
```

```python
#4 Create a NumPy array from Survived and find the index positions of all passengers who survived. 

survived = df['Survived'].to_numpy()

print(np.where(survived == 1))
```

```python
#5 
"""
Create a NumPy matrix containing these columns in this exact order: 
Pclass, Age, SibSp, Parch, Fare 

Then display its shape. 
"""

plcass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()

mat = np.column_stack((plcass, age, sibsp, parch, fare))

mat.shape
```

```python
#6 Using the matrix from Q5, split it vertically into two equal parts using NumPy. 

print(np.array_split(mat,2))
```

```python
#7 Create separate NumPy arrays for Pclass, SibSp, and Parch, then join them horizontally into a single feature matrix using an appropriate NumPy joining method

mat = np.column_stack((pclass, sibsp, parch))

mat
```

```python
#8 Increase every passenger's Fare by 10% using NumPy broadcasting, without using a loop. 

increase = fare + (fare * 10) / 100

print(increase)
```

```python
#9 Create a NumPy Age array and replace every missing Age value with the overall mean Age, using NumPy vectorized operations.

age_mean = np.nanmean(age)

age[np.isnan(age)] = age_mean

print(age)
```

```python
#10
"""
Create a NumPy matrix containing: 
Survived, Pclass, Age, SibSp, Parch, Fare 
 
Then standardize the Fare column using vectorized NumPy operations so that you 
transform it using its mean and standard deviation.
"""

fare_mean = fare.mean()
fare_std = fare.std()

fare_standardized = (fare - fare_mean) / fare_std

print(fare_standardized)
```
