### Q1 - Passenger ID

```python
print(f"Shape : {df['PassengerId'].shape}")
print(f"ndim : {df['PassengerId'].ndim}")
print(f"minimum ID : {df['PassengerId'].min()}")
print(f"maximum ID : {df['PassengerId'].max()}")
print(f"first 10 IDs : {df['PassengerId'][:10]}")
```

### Q2 - Fare Analysis

```python
print(f"average fare : {df['Fare'].mean()}")
print(f"highest fare : {df['Fare'].max()}")
print(f"lowest fare : {df['Fare'].min()}")
print(f"first 10 fare : {df['Fare'][:10]}")

add = df['Fare'] + 10

print(f"new average : {add.mean()}")
```

### Q3 - Age Analysis 

```python
print(f"average age : {df['Age'].mean()}")
print(f"highest age : {df['Age'].max()}")
print(f"lowest age : {df['Age'].min()}")
print(f"first 10 age : {df['Age'][:10]}")

if (df['Age'] > 80).any():
    print("Yes, passenger is older than 80")
else:
    print("No, passenger is not older than 80")
```

### Q4 - Passenger Class 

```python
print(f"minimum class : {df['Pclass'].min()}")
print(f"maximum class : {df['Pclass'].max()}")
print(f"first 20 values : {df['Pclass'][:20]}")
print(f"count passengers in Class 1, 2, and 3 : {df['Pclass'].value_counts()}")

if ((df['Pclass'] >= 1) & (df['Pclass'] <= 3)).all():
    print("Yes all values are between 1 and 3")
else:
    print("No all values are not between 1 and 3")
```

### Q5 - Survival Analysis 

```python
print(f"total survivors : {(df['Survived'] == 1).sum()}")
print(f"total non-survivors : {(df['Survived'] == 0).sum()}")
print(f"survival rate : {(df['Survived'] == 1).sum() / (len(df['Survived'])) * 100.0}")
print(f"first 20 values  : {df['Survived'][:20]}")

if (df['Survived'] == 1).any():
    print("Yes, at least one passenger survived")
else:
    print("No passenger survived")
```
