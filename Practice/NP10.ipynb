### Q1 - Passenger ID

```python
print(f"Shape : {df['PassengerId'].shape}")
print(f"ndim : {df['PassengerId'].ndim}")
print(f"minimum ID : {df['PassengerId'].min()}")
print(f"maximum ID : {df['PassengerId'].max()}")
print(f"first 10 IDs : {df['PassengerId'][:10]}")
```

### Q2 - Fare Analysis

```python
print(f"average fare : {df['Fare'].mean()}")
print(f"highest fare : {df['Fare'].max()}")
print(f"lowest fare : {df['Fare'].min()}")
print(f"first 10 fare : {df['Fare'][:10]}")

add = df['Fare'] + 10

print(f"new average : {add.mean()}")
```

### Q3 - Age Analysis 

```python
print(f"average age : {df['Age'].mean()}")
print(f"highest age : {df['Age'].max()}")
print(f"lowest age : {df['Age'].min()}")
print(f"first 10 age : {df['Age'][:10]}")

if (df['Age'] > 80).any():
    print("Yes, passenger is older than 80")
else:
    print("No, passenger is not older than 80")
```

### Q4 - Passenger Class 

```python
print(f"minimum class : {df['Pclass'].min()}")
print(f"maximum class : {df['Pclass'].max()}")
print(f"first 20 values : {df['Pclass'][:20]}")
print(f"count passengers in Class 1, 2, and 3 : {df['Pclass'].value_counts()}")

if ((df['Pclass'] >= 1) & (df['Pclass'] <= 3)).all():
    print("Yes all values are between 1 and 3")
else:
    print("No all values are not between 1 and 3")
```

### Q5 - Survival Analysis 

```python
print(f"total survivors : {(df['Survived'] == 1).sum()}")
print(f"total non-survivors : {(df['Survived'] == 0).sum()}")
print(f"survival rate : {(df['Survived'] == 1).sum() / (len(df['Survived'])) * 100.0}")
print(f"first 20 values  : {df['Survived'][:20]}")

if (df['Survived'] == 1).any():
    print("Yes, at least one passenger survived")
else:
    print("No passenger survived")
```

### Q6 - Family Matrix 

```python
"""
Create a NumPy matrix using: 
SibSp and Parch 
 
Find: 
 shape  
 dimensions  
 first 10 rows  
 maximum individual value  
 total of each column 
"""

family = df[['SibSp','Parch']]

print(f"shape : {family.shape}")
print(f"ndim : {family.ndim}")
print(f"first 10 rows : {family[:10]}")
print(f"maximum individual value : {family.max()}")
print(f"total of each column : {family.sum()}")
```

### Q7 - Passenger Matrix

```python
"""
Create a matrix using: 
PassengerId, Survived, Pclass, Age, Fare 
 
Find: 
 shape  
 first 5 passengers  
 Fare column  
 highest individual Fare  
 lowest individual Fare 
"""

passenger = df[['PassengerId','Survived','Pclass','Age','Fare']]

print(f"shape : {passenger.shape}")
print(f"first 5 passengers : {passenger[:5]}")
print(f"Fare column : {passenger['Fare']}")
print(f"highest individual Fare : {passenger['Fare'].max()}")
print(f"lowest individual Fare : {passenger['Fare'].min()}")
```

### Q8 - Reshaping

```python
"""
Take the first 20 rows of the matrix from Q7. 
 Display its original shape.  
 Reshape it into 10 × 10.  
 Display the reshaped matrix.  
 Extract the second row.  
 Extract the third column.  
"""

mat = np.array(passenger[:20])

print(f"shape : {mat.shape}")

re = mat.reshape(10,10)

print(f"reshape : {re}")

print(f"second row : {re[1]}")
print(f"third column : {re[:,2]}")
```

### Q9 - Flattening & Transpose

```python
"""
Using the reshaped matrix from Q8: 
 flatten it  
 display its shape  
 transpose the original Q8 matrix  
 display the transposed shape  
 find the highest individual value  
"""

mat = np.array(passenger[:20]).reshape(10,10)

print(f"flatten : {mat.flatten()}")
print(f"shape : {mat.shape}")

transposed = mat.transpose()
print(f"transposed shape : {transposed.shape}")

print(f"highest individual value : {mat.max()}")
```

### Q10 - Final Data Analyst Challenge
