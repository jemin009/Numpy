```python
import numpy as np
```

```python
#1
"""
Perform: 
1. Create a NumPy array.  
2. Find the total visitors.  
3. Find the average visitors.  
4. Find the maximum visitors.  
5. Extract visitors from Day 3 to Day 7.  
6. Extract the last 3 days.  
7. Increase every day's visitors by 5%.  
8. Find the new maximum.  
9. Check whether the new maximum exceeds ₹ — ignore currency; treat it as 
visitor count — 1900.  
10. Print the shape and ndim.  
"""

visitors = [1250, 1380, 1425, 1190, 1560, 1680, 1490, 1720, 1810, 1650] 

arr = np.array(visitors)

print(f"total visitors : {arr.sum()}")
print(f"average visitors : {arr.mean()}")
print(f"maximum visitors : {arr.max()}")
print(f"visitors from Day 3 to Day 7 : {arr[2:7]}")
print(f"last 3 days : {arr[-3:]}")

increased = arr + (arr * 5) / 100
print(f"new maximum : {increased.max()}")

if increased.max() > 1900:
    print("Yes, the new maximum exceeds 1900")
else:
    print("No, the new maximum not exceeds 1900")

print(f"Shape : {increased.shape}")
print(f"Ndim : {increased.ndim}")
```

```python
#2
"""
Using np.arange(): 
1. Create targets from ₹20,000 up to ₹50,000.  
2. Find the average target.  
3. Find the maximum target.  
4. Extract the first 3 targets.  
5. Extract the last 2 targets.  
6. Increase all targets by 10%.  
7. Find the new maximum. 

A company wants to create monthly sales targets starting from ₹20,000 and 
increasing by ₹5,000. 
"""

sales = np.arange(20000,50001,5000)

print(sales)
print(f"average target : {sales.mean()}")
print(f"maximum target : {sales.max()}")
print(f"first 3 targets : {sales[:3]}")
print(f"last 2 targets : {sales[-2:]}")

increased = sales + (sales * 10) / 100
print(f"new maximum : {increased.max()}")
```

```python
#3
"""
Using np.linspace(): 
1. Create the benchmark array.  
2. Find its shape.  
3. Find its ndim.  
4. Find the average benchmark.  
5. Find the maximum and minimum.  
6. Extract the first 3 benchmarks.  
7. Add 5 points to every benchmark.  
8. Find the new maximum. 

An HR team wants to generate 6 evenly spaced performance benchmark scores 
between 60 and 90, including both endpoints. 
"""

benchmark = np.linspace(60,90,6)

print(benchmark)
print(f"Shape : {benchmark.shape}")
print(f"Ndim : {benchmark.ndim}")
print(f"average benchmark : {benchmark.mean()}")
print(f"maximum benchmark : {benchmark.max()}")
print(f"minimum benchmark : {benchmark.min()}")
print(f"first 3 benchmarks : {benchmark[:3]}")

add = benchmark + 5

print(f"New maximum : {add.max()}")
```

```python
#4
"""
Perform: 
1. Create the 2D array.  
2. Extract the Monthly Spend column.  
3. Extract the Orders column.  
4. Extract the first 4 customers.  
5. Extract Age and Orders for all customers.  
6. Find total Monthly Spend.  
7. Find average Orders.  
8. Find the highest individual Monthly Spend.  
9. Add 500 to every Monthly Spend.  
10. Display the updated customer data.  

Columns: 
Age → Monthly Spend → Orders
"""

customers = [ 
    [21, 4500, 5], 
    [28, 7200, 8], 
    [35, 6800, 7], 
    [42, 9200, 10], 
    [31, 5600, 6], 
    [25, 8100, 9] 
]

arr = np.array(customers)

print(f"Monthly Spend : {arr[:,1]}")
print(f"Orders : {arr[:,-1]}")
print(f"first 4 customers : {arr[:4]}")
print(f"Age and Orders for all customers : {arr[:,::2]}")
print(f"total Monthly Spend : {arr[:,1].sum()}")
print(f"average Orders : {arr[:,-1].mean()}")
print(f"highest individual Monthly Spend : {arr[:,1].max()}")

arr[:,1] = arr[:,1] + 500

print(f"updated customer data : {arr}")
```

```python
#5
"""
Perform: 
1. Find shape and ndim.  
2. Find total sales of each store.  
3. Find average sales of each store.  
4. Find total sales of each quarter.  
5. Extract Q2 and Q3 for all stores.  
6. Extract all quarters of Store 3.  
7. Find the highest individual sales value.  
8. Find the lowest individual sales value.  
9. Increase all sales by 8%.  
10. Find the new maximum.   

Rows = Stores 
Columns = Q1, Q2, Q3, Q4 
"""

sales = [ 
    [12000, 15000, 18000, 21000], 
    [14000, 17000, 16000, 23000], 
    [11000, 13000, 19000, 22000], 
    [16000, 18000, 21000, 25000] 
]

arr = np.array(sales)

print(f"Shape : {arr.shape}")
print(f"Ndim : {arr.ndim}")
print(f"average sales of each store : {arr.mean(axis=1)}")
print(f"total sales of each quarter : {arr.sum(axis=0)}")
print(f"Q2 and Q3 for all stores : {arr[:,1:3]}")
print(f"all quarters of Store 3 : {arr[2]}")
print(f"highest individual sales value : {arr.max(axis=1)}")
print(f"lowest individual sales value : {arr.min(axis=1)}")

increased = arr + (arr * 8) / 100

print(f"new maximum : {increased.max()}")
```

```python
#8
"""
Perform: 
1. Create it using np.identity().  
2. Create another 4 × 4 identity matrix using np.eye().  
3. Check their shapes.  
4. Check their ndims.  
5. Find the maximum value of each matrix.  
6. Find the minimum value of each matrix.  
7. Explain from your output whether both matrices contain the same values. 

A Data Science team wants to create an identity matrix of size 4 × 4. 
"""

identity = np.identity(4)
eye = np.eye(4)

print(f"identity : {identity}")
print(f"eye : {eye}")

print(f"Shape of identity : {identity.shape}")
print(f"Shape of eye : {eye.shape}")

print(f"Ndim of identity : {identity.ndim}")
print(f"Ndim of eye : {eye.ndim}")

print(f"maximum value of identity : {identity.max()}")
print(f"maximum value of eye : {eye.max()}")

print(f"minimum value of identity : {identity.min()}")
print(f"minimum value of eye : {eye.min()}")

print("Both matrices contain the same values: Yes")
```
