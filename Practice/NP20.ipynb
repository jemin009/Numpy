```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1 
"""
Create: 
Pclass_Avg_Fare 
using groupby() + transform(). 
 
Then create: 
Fare_Difference 
 
using: 
Fare - Pclass_Avg_Fare 
 
Display: 
Name, Pclass, Fare, Pclass_Avg_Fare, Fare_Difference 
for the first 10 passengers.
"""

df['Pclass_Avg_Fare'] = df.groupby('Pclass')['Fare'].transform('mean')

df['Fare_Difference'] = df['Fare'] - df['Pclass_Avg_Fare']

show = df[['Name','Pclass','Fare','Pclass_Avg_Fare','Fare_Difference']].head(10)

print(show)
```

```python
#2 
"""
Create a pivot table showing the survival rate for each combination of: 
 Rows → Pclass  
 Columns → Sex  
Display the result as percentages. 
"""

pvt = df.pivot_table(
    index = 'Pclass',
    columns = 'Sex',
    values = 'Survived',
    aggfunc = 'mean'
) * 100

print(pvt)
```

```python
#3 
"""
Find passengers whose Name contains either: 
 "Mr."  
 "Dr."  
 
Display: 
Name, Sex, Age, Pclass, Survived
"""

mrd = df['Name'].str.contains('Mr.') | df['Name'].str.contains('Dr.')

show = df.loc[mrd, ['Name','Sex','Age','Pclass','Survived']]

print(show)
```

```python
#4 
"""
Create: 
Fare_Adjusted 
 
Rules: 
 Fare < 50 → increase by 20%  
 Fare 50–100 → increase by 10%  
 Fare > 100 → increase by 5%  
Display the 10 passengers with the highest Fare_Adjusted.
"""

df['Fare_Adjusted'] = np.where(
    df['Fare'] < 50,
    df['Fare'] * 1.20,
    
    np.where(
        (df['Fare'] >= 50) & (df['Fare'] <= 100),
        df['Fare'] * 1.10,
        df['Fare'] * 1.05
    )
)

show = df.sort_values('Fare_Adjusted', ascending=False).head(10)

print(show)
```

```python
#5 
"""
Create a separate DataFrame containing: 
Pclass 
Average_Age 
Average_Fare 
Then merge it back into the Titanic DataFrame using Pclass. 
 
Display: 
Name, Pclass, Age, Fare, Average_Age, Average_Fare 
for the first 10 passengers.
"""

summary = df.groupby('Pclass').agg(
    Average_Age=('Age', 'mean'),
    Average_Fare=('Fare', 'mean')
).reset_index()

result = df.merge(summary, on='Pclass')

show = result[['Name', 'Pclass', 'Age', 'Fare', 'Average_Age', 'Average_Fare']].head(10)

print(show)
```

```python
#6 
"""
Create a NumPy Age array. 
Replace missing Age values with the median Age of the corresponding Pclass. 
 
Create: 
Age_Filled_By_Class 
 
Display 10 passengers who originally had missing Age: 
Name, Pclass, Age, Age_Filled_By_Class
"""

age = df['Age'].to_numpy()

df['Age_Filled_By_Class'] = df.groupby('Pclass')['Age'].transform('median')

show = df.loc[np.isnan(age),['Name','Pclass','Age','Age_Filled_By_Class']].head(10)

print(show)
```

```python
#7 
"""
For each Sex, calculate: 
 Minimum Fare  
 Maximum Fare  
 Average Fare  
 Median Age  
 Survival Rate  
Use a single groupby().agg() operation.
"""

grp = df.groupby('Sex').agg(
                                MinFare = ('Fare','min'),
                                MaxFare = ('Fare','max'),
                                AvgFare = ('Fare','mean'),
                                MedAge = ('Age','median'),
                                SurvivalRate = ('Survived','mean')
)

print(grp)
```

```python
#8 
"""
Create: 
FamilySize = SibSp + Parch + 1 
FarePerPerson = Fare / FamilySize 
 
Then create a NumPy matrix containing: 
Pclass 
Age 
FamilySize 
Fare 
FarePerPerson 
Replace missing Age with the median Age. 
 
Print: 
1. Matrix shape  
2. First 5 rows  
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

df['FarePerPerson'] = df['Fare'] / df['FamilySize']

pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
familysize = df['FamilySize'].to_numpy()
fare = df['Fare'].to_numpy()
fareperperson = df['FarePerPerson'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

mat = np.column_stack((pclass,age, familysize,fare,fareperperson))

print(mat.shape)

print(mat[:5])
```

```python
#9 
"""
Create: 
Fare_Percent_Of_Class_Avg 
 
using: 
(Fare / Pclass_Avg_Fare) × 100 
where Pclass_Avg_Fare is created using groupby() + transform(). 
Then find the 5 passengers whose Fare is furthest above their Pclass average, using 
the largest positive percentage. 

Display: 
Name, Pclass, Fare, Pclass_Avg_Fare, Fare_Percent_Of_Class_Avg
"""

df['Pclass_Avg_Fare'] = df.groupby('Pclass')['Fare'].transform('mean')

df['Fare_Percent_Of_Class_Avg'] = (df['Fare'] / df['Pclass_Avg_Fare']) * 100

show = (
    df.sort_values('Fare_Percent_Of_Class_Avg', ascending=False)
      [['Name', 'Pclass', 'Fare','Pclass_Avg_Fare', 'Fare_Percent_Of_Class_Avg']]
      .head(5)
)

print(show)
```

```python
#10
"""
Create a final DataFrame containing: 
Survived 
Pclass 
Age 
SibSp 
Parch 
Fare 
FamilySize 
FarePerPerson 
PclassAvgFare 
FarePercentOfClassAvg 
 
Requirements: 
1. Replace missing Age with the median Age.  
2. Create FamilySize.  
3. Create FarePerPerson.  
4. Create PclassAvgFare using groupby() + transform().  
5. Create FarePercentOfClassAvg.  
6. Min-max normalize Fare.  
7. Standardize FarePerPerson.  
8. Convert the final features to a NumPy matrix.  
9. Transpose the matrix.  
10. Print the final matrix shape.
"""

age = df['Age'].to_numpy()

# 1. Replace missing Age with median
df['Age'] = np.where(np.isnan(age), np.nanmedian(age), age)

# 2. FamilySize
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# 3. FarePerPerson
df['FarePerPerson'] = df['Fare'] / df['FamilySize']

# 4. PclassAvgFare
df['PclassAvgFare'] = (df.groupby('Pclass')['Fare'].transform('mean'))

# 5. FarePercentOfClassAvg
df['FarePercentOfClassAvg'] = (df['Fare'] / df['PclassAvgFare']) * 100

# 6. Min-max normalize Fare
fare_min = df['Fare'].min()
fare_max = df['Fare'].max()

df['Fare'] = ((df['Fare'] - fare_min) /(fare_max - fare_min))

# 7. Standardize FarePerPerson
df['FarePerPerson'] = ((df['FarePerPerson'] - df['FarePerPerson'].mean()) /df['FarePerPerson'].std())

# 8. Convert to NumPy matrix
matrix = df[
    [
        'Survived', 'Pclass', 'Age', 'SibSp', 'Parch','Fare', 'FamilySize', 'FarePerPerson',
        'PclassAvgFare', 'FarePercentOfClassAvg'
    ]
].to_numpy()

# 9. Transpose
trans = matrix.T 

# 10. Final shape
print(trans.shape)
```
