```python
#1 Load the Titanic dataset and inspect its number of rows and columns. Then convert the Fare column into a NumPy array and find its mean Fare.

print(df.shape)

fare = df['Fare'].to_numpy()

print(np.mean(fare))
```

```python
#3 Create a NumPy array from Fare and find the indices of the 10 highest fares. Use those indices to display the corresponding passengers' Name and Fare from the DataFrame. 

fare = df['Fare'].to_numpy()

highest = np.argsort(fare)[-10:][::-1]

show = df.loc[highest,['Name','Fare']]

show
```

```python
#5 
"""
Using Pandas, find passengers who satisfy all of these conditions: 
 Age is greater than 30  
 Fare is greater than 100  
 Passenger survived  
Display their Name, Age, Fare, and Pclass.
"""

filter = df.loc[(df['Age'] > 30) & (df['Fare'] > 100) & (df['Survived'] == 1),['Name','Age','Fare','Pclass']]

print(filter)
```

```python
#6 Check the data type of Fare. Convert Fare to a numeric data type if necessary, create a NumPy array from it, and find the minimum, maximum, and mean Fare.

fare = df['Fare'].to_numpy()

print(df['Fare'].dtype)

print(np.min(fare))
print(np.max(fare))
print(np.mean(fare))
```

```python
#7 Using Pandas, select Pclass, Age, SibSp, Parch, and Fare. Replace missing Age values with the median Age, convert the selected data into a NumPy matrix, and display its shape. 

median = np.nanmedian(df['Age'])

selected = df[['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].copy()

selected['Age'] = selected['Age'].fillna(median)

matrix = selected.to_numpy()

print(matrix)
print(matrix.shape)
```

```python
#8 Create a NumPy array from Fare. Increase fares above 200 by 5% and fares of 200 or below by 10%. Store the transformed values back into the DataFrame as a new column called AdjustedFare.

fare = df['Fare'].to_numpy()

increased = np.where(fare > 200, fare + (fare * 5) / 100, fare + (fare * 10) / 100)

df['AdjustedFare'] = increased

print(df[['Fare', 'AdjustedFare']])
```

```python
#9 Create a NumPy matrix containing Pclass, Age, SibSp, Parch, and Fare. Replace missing Age values with the median Age and sort the complete matrix by Fare from highest to lowest.

pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()

median = np.nanmedian(age)

age = np.where(np.isnan(age),median,age)

mat = np.column_stack((pclass,age,sibsp,parch,fare))

sort = mat[np.argsort(mat[:, 4])[::-1]]

print(sort)
```

```python
#10 Using Pandas and NumPy, create a feature matrix containing Survived, Pclass, Age, SibSp, Parch, and Fare. Handle missing Age using the median, min-max normalize Fare, and transpose the final NumPy matrix.

median = np.nanmedian(df['Age'])

selected = df[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].copy()

selected['Age'] = selected['Age'].fillna(median)

fare_min = selected['Fare'].min()
fare_max = selected['Fare'].max()

selected['Fare'] = ((selected['Fare'] - fare_min) /(fare_max - fare_min))

matrix = selected.to_numpy()

final_matrix = matrix.T

print(final_matrix)
```
