```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1 Using the Fare NumPy array, find the 5 most common fare values and their frequencies using NumPy.

fare = df['Fare'].to_numpy()

unique, counts = np.unique(fare, return_counts=True)

top5 = np.argsort(counts)[-5:][::-1]

print("Top 5 most common fares:")
print(unique[top5])

print("Their frequencies:")
print(counts[top5])
```

```python
#2 Using the Age NumPy array, find the minimum, maximum, and mean age, while ignoring missing values.

age = df['Age'].to_numpy()

print(f"Min = {np.nanmin(age)}")
print(f"Max = {np.nanmax(age)}")
print(f"Mean = {np.nanmean(age)}")
```

```python
#3 Using the Pclass NumPy array, find the number of passengers in each unique passenger class using NumPy. 

pclass = df['Pclass'].to_numpy()

classes, counts = np.unique(pclass, return_counts=True)

print(classes)
print(counts)
```

```python
#4 Using Survived and Pclass, create a NumPy matrix and calculate the total number of survivors for each passenger class.

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()

mat = np.column_stack((survived, pclass))

print(np.sum((survived == 1) & (pclass == 1)))
print(np.sum((survived == 1) & (pclass == 2)))
print(np.sum((survived == 1) & (pclass == 3)))
```

```python
#5 Using the Fare NumPy array, find the indices of passengers whose fare is above the overall mean fare.

fare = df['Fare'].to_numpy()

print(np.where(fare > fare.mean()))
```

```python
#6 Using Age and Pclass, create a NumPy matrix containing both features and sort the rows by Age.

age = df['Age'].to_numpy()
pclass = df['Pclass'].to_numpy()

mat = np.column_stack((age, pclass))

sorted_mat = mat[np.argsort(mat[:, 0])]

print(sorted_mat)
```

```python
#7 Using the Fare array, create a new NumPy array where fares greater than 100 are replaced with 100, while all other fares remain unchanged.

fare = df['Fare'].to_numpy()

print(np.where(fare > 100, 100, fare))
```

```python
#8 Using Age, replace missing values with the median age instead of the mean, using NumPy vectorized operations. 

age = df['Age'].to_numpy()

median_age = np.nanmedian(age)

age[np.isnan(age)] = median_age

print(age)
```

```python
#9 Create a NumPy feature matrix containing Pclass, Age, SibSp, Parch, and Fare. Standardize the Age and Fare columns using vectorized NumPy operations. 

# Convert columns to NumPy arrays
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy().copy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()

# Replace missing Age values with the median Age
age[np.isnan(age)] = np.nanmedian(age)

# Create the feature matrix
mat = np.column_stack((pclass, age, sibsp, parch, fare))

# Standardize Age
age_mean = np.mean(mat[:, 1])
age_std = np.std(mat[:, 1])
mat[:, 1] = (mat[:, 1] - age_mean) / age_std

# Standardize Fare
fare_mean = np.mean(mat[:, 4])
fare_std = np.std(mat[:, 4])
mat[:, 4] = (mat[:, 4] - fare_mean) / fare_std

print(mat)
```

```python
#10 Create a NumPy feature matrix containing Survived, Pclass, Age, SibSp, Parch, and Fare. Normalize the Fare column using min-max normalization so that its values are between 0 and 1.

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy().copy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()

# Replace missing Age values
age[np.isnan(age)] = np.nanmedian(age)

# Create feature matrix
mat = np.column_stack((survived, pclass, age, sibsp, parch, fare))

# Min-max normalization of Fare
fare_min = np.min(mat[:, -1])
fare_max = np.max(mat[:, -1])

mat[:, -1] = (mat[:, -1] - fare_min) / (fare_max - fare_min)

print(mat)
```
