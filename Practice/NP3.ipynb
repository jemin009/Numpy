```python
#1
"""
Using NumPy, calculate: 
1. Total sales  
2. Average weekly sales  
3. Highest weekly sales  
4. Lowest weekly sales  
5. Determine whether the average weekly sales are above 15,000
"""

import numpy as np

sales = [12500, 14800, 11200, 16500, 17800, 13900, 19200, 15400] 

arr = np.array(sales)

print(f"Total sales : {arr.sum()}")
print(f"Average weekly sales : {arr.mean()}")
print(f"Highest weekly sales : {arr.max()}")
print(f"Lowest weekly sales : {arr.min()}")

if arr.mean() > 15000:
    print("Yes, average weekly sales are above 15,000")
else:
    print("No, average weekly sales are not above 15,000")

```

```python
#8
"""
For both products: 
1. Calculate total revenue  
2. Calculate average revenue  
3. Find highest weekly revenue  
4. Find lowest weekly revenue  
5. Determine which product generated the higher total revenue  
"""

product_A = [12000, 14500, 13200, 15800, 17100] 
product_B = [13500, 12800, 14900, 16200, 15500]

arrA = np.array(product_A)
arrB = np.array(product_B)

print(f"Total revenue of Product_A : {arrA.sum()}")
print(f"Total revenue of Product_B : {arrB.sum()}")

print(f"Average revenue of Product_A : {arrA.mean()}")
print(f"Average revenue of Product_B : {arrB.mean()}")

print(f"Highest weekly revenue of Product_A : {arrA.max()}")
print(f"Highest weekly revenue of Product_B : {arrB.max()}")

print(f"Lowest weekly revenue of Product_A : {arrA.min()}")
print(f"Lowest weekly revenue of Product_B : {arrB.min()}")

if arrA.sum() > arrB.sum():
    print("Product_A generated the higher total revenue")
else:
    print("Product_B generated the higher total revenue")
```

```python
#9
"""
Using NumPy: 
1. Create a NumPy array  
2. Increase every value by 50  
3. Multiply every value by 2  
4. Calculate the total of the final values  
5. Calculate the average of the final values  
6. Find the highest final value  
"""

import numpy as np

sales = [100, 200, 300, 400, 500]

arr = np.array(sales)

increase = arr + 50
final_values = increase * 2

print(f"Final values : {final_values}")
print(f"Total : {final_values.sum()}")
print(f"Average : {final_values.mean()}")
print(f"Highest value : {final_values.max()}")
```
