```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1
"""
For each Sex, calculate: 
 Average Fare  
 Median Fare  
 Average Age  
 Survival Rate  
Use one groupby().agg() operation.
"""

grp = df.groupby('Sex').agg(
                    AvgFare = ('Fare','mean'),
                    MedianFare = ('Fare','median'),
                    AvgAge = ('Age','mean'),
                    SurvivalRate = ('Survived','mean')
).reset_index()

print(grp)
```

```python
#2
"""
Create Pclass_Avg_Age using groupby() + transform(). 

Then create: 
Age_Difference = Age - Pclass_Avg_Age 

Display the 10 passengers with the largest absolute Age_Difference. 
"""

df['Pclass_Avg_Age'] = df.groupby('Pclass')['Age'].transform('mean')

df['Age_Difference'] = (df['Age'] - df['Pclass_Avg_Age']).abs()

sort = df.sort_values('Age_Difference', ascending=False).head(10)

print(sort)

```

```python
#3
"""
Create a pivot table showing average Fare: 
 Rows → Pclass  
 Columns → Embarked  
 Values → Fare  
Use mean as the aggregation function.
"""

pt = df.pivot_table(
        index = 'Pclass',
        columns = 'Embarked',
        values = 'Fare',
        aggfunc = 'mean'
)

print(pt)
```

```python
#4
"""
Find passengers whose names contain either "Mr." or "Dr.". 

Display: 
Name, Sex, Age, Pclass, Fare, Survived
"""

mrd = df['Name'].str.contains('Mr.') | df['Name'].str.contains('Dr.')

show = df.loc[mrd,['Name','Sex','Age','Pclass','Fare','Survived']]

print(show)
```

```python
#5 
"""
Create AdjustedFare using these rules: 
 Fare < 30 → increase by 20%  
 Fare 30–100 → increase by 10%  
 Fare > 100 → increase by 5%  

Display the 10 passengers with the highest AdjustedFare. 
"""

df['AdjustedFare'] = np.where(
                                df['Fare'] < 30, df['Fare'] + (df['Fare'] * 0.20),
                                np.where(df['Fare'].between(30,100), df['Fare'] + (df['Fare'] * 0.10),
                                         df['Fare'] + (df['Fare'] * 0.05))
                                )

sort = df.sort_values('AdjustedFare', ascending=False).head(10)

print(sort)
```

```python
#6
"""
Create a separate summary DataFrame containing, for each Pclass: 
 Average_Fare  
 Median_Fare  
 Survival_Rate  
Then merge this summary back into the original Titanic DataFrame using Pclass. 

Display the first 10 rows with: 
Name, Pclass, Fare, Average_Fare, Median_Fare, Survival_Rate
"""

summary = df.groupby('Pclass').agg(
                                        Average_Fare = ('Fare','mean'),
                                        Median_Fare = ('Fare','median'),
                                        Survival_Rate = ('Survived','mean')
).reset_index()

join = df.merge(summary, on='Pclass')

show = join[['Name','Pclass','Fare','Average_Fare','Median_Fare','Survival_Rate']].head(10)

print(show)
```

```python
#7
"""
Create a NumPy array from Fare. 
Using NumPy: 
1. Find the minimum Fare.  
2. Find the maximum Fare.  
3. Find the mean Fare.  
4. Create a normalized Fare array using min-max normalization.  

Do not use a Pandas normalization method.
"""

fare = df['Fare'].to_numpy()

min_fare = fare.min()
max_fare = fare.max()
mean_fare = fare.mean()

normalized_fare  = (fare - min_fare) / (max_fare - min_fare)

print(normalized_fare)
```

```python
#8
"""
Create: 
FamilySize = SibSp + Parch + 1 

Then create: 
FarePerPerson = Fare / FamilySize 

Create a NumPy matrix containing: 
Pclass 
FamilySize 
Fare 
FarePerPerson 
Survived 

Print: 
 Matrix shape  
 First 5 rows 
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['FarePerPerson'] = df['Fare'] / df['FamilySize']

pclass = df['Pclass'].to_numpy()
familysize = df['FamilySize'].to_numpy()
fare = df['Fare'].to_numpy()
fareperperson = df['FarePerPerson'].to_numpy()
survived = df['Survived'].to_numpy()

mat = np.column_stack((pclass,familysize,fare,fareperperson,survived))

print(mat.shape)

print(mat[:5])
```

```python
#9
"""
Create: 
Pclass_Avg_Fare 
Fare_Percent_Of_Class_Avg 

where: 
Fare_Percent_Of_Class_Avg = (Fare / Pclass_Avg_Fare) × 100 

Then display the 5 passengers whose Fare is furthest below their own Pclass 
average.
"""

df['Pclass_Avg_Fare'] = df.groupby('Pclass')['Fare'].transform('mean')

df['Fare_Percent_Of_Class_Avg'] = (df['Fare'] / df['Pclass_Avg_Fare']) * 100

show = df[df['Fare'] < df['Pclass_Avg_Fare']].sort_values('Fare_Percent_Of_Class_Avg').head(5)

print(show)
```

```python
#10
"""
Create these final features: 
Survived 
Pclass 
Age 
FamilySize 
Fare 
FarePerPerson 
PclassAvgFare 
FarePercentOfClassAvg 

Requirements: 
1. Replace missing Age with the median Age.  
2. Create FamilySize, FarePerPerson, PclassAvgFare, and FarePercentOfClassAvg.  
3. Min-max normalize Fare and standardize FarePerPerson.  
4. Convert these final features to a NumPy matrix and transpose it.  
5. Print the final matrix shape.
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['FarePerPerson'] = df['Fare'] / df['FamilySize']
df['PclassAvgFare'] = df.groupby('Pclass')['Fare'].transform('mean')
df['FarePercentOfClassAvg'] = (df['Fare'] / df['PclassAvgFare']) * 100

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
familysize = df['FamilySize'].to_numpy()
fare = df['Fare'].to_numpy()
fareperperson = df['FarePerPerson'].to_numpy()
pclassavgfare = df['PclassAvgFare'].to_numpy()
farepercentofclassavg = df['FarePercentOfClassAvg'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

fare = (fare - fare.min()) / (fare.max() - fare.min())

fareperperson = (fareperperson - fareperperson.mean()) / fareperperson.std()

mat = np.column_stack((survived, pclass, age, familysize, fare, fareperperson, pclassavgfare, farepercentofclassavg))

trans = mat.T

print(trans.shape)
```
