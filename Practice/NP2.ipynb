```python
#8 
"""
Using NumPy: 
1. Calculate total January revenue.  
2. Calculate total February revenue.  
3. Calculate average January revenue.  
4. Calculate average February revenue.  
5. Identify which month had higher total revenue. 
"""

january = [1200, 1500, 1350, 1800, 1650] 
february = [1400, 1550, 1600, 1900, 1750]

jan_total = np.array(january).sum()
feb_total = np.array(february).sum()

jan_avg = np.array(january).mean()
feb_avg = np.array(february).mean()

print(f"Total January revenue : {jan_total}")
print(f"Total February revenue : {feb_total}")
print(f"Average January revenue : {jan_avg}")
print(f"Average February revenue : {feb_avg}")

if jan_total > feb_total:
    print("January month had higher total revenue")
else:
    print("February month had higher total revenue")
```
