```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1 
"""
For each Pclass, calculate: 
 Average Fare  
 Median Fare  
 Average Age  
 Survival Rate 
"""

grp = df.groupby('Pclass').agg(
                                AvgFare = ('Fare','mean'),
                                MedianFare = ('Fare','median'),
                                AvgAge = ('Age','mean'),
                                SurvivalRate = ('Survived','mean')
)
print(grp)
```

```python
#2
"""
Create Pclass_Avg_Fare using groupby() + transform(). 
 
Then create Fare_Difference: 
Fare - Pclass_Avg_Fare 
Display the 10 passengers with the largest positive difference.
"""

df['Pclass_Avg_Fare'] = df.groupby('Pclass')['Fare'].transform('mean')

df['Fare_Difference'] = df['Fare'] - df['Pclass_Avg_Fare']

sort = df.sort_values('Fare_Difference',ascending=False).head(10)

print(sort)
```

```python
#3
"""
Create a pivot table showing average Fare: 
 Rows → Pclass  
 Columns → Sex 
"""

pt = df.pivot_table(
        index = 'Pclass',
        columns = 'Sex',
        values = 'Fare',
        aggfunc = 'mean'
)

print(pt)
```

```python
#4
"""
Find passengers whose names contain either "Miss." or "Mrs.". 
 
Display: 
Name, Sex, Age, Fare, Survived
"""

mrs = df['Name'].str.contains('Miss.') | df['Name'].str.contains('Mrs.')

show = df.loc[mrs,['Name','Sex','Age','Fare','Survived']]

print(show)
```

```python
#5
"""
Create AdjustedFare: 
 Fare ≤ 50 → increase by 15%  
 Fare > 50 → increase by 5%  
Display the 10 highest AdjustedFare values.
"""

df['AdjustedFare'] = np.where(df['Fare'] <= 50,
                      df['Fare'] + (df['Fare'] * 0.15),
                      df['Fare'] + (df['Fare'] * 0.05))

show = df.sort_values('AdjustedFare',ascending=False).head(10)

print(show)
```

```python
#6
"""
Create a NumPy Fare array and create: 
Fare_Category 
 
using: 
 Fare < 20 → "Low"  
 Fare 20–100 → "Medium"  
 Fare > 100 → "High"  
 
Display: 
Name, Fare, Fare_Category 
for the first 15 passengers.
"""

fare = df['Fare'].to_numpy()

df['Fare_Category'] = np.where(fare < 20,'Low',
                        np.where((fare >= 20) & (fare <= 100),'Medium','High')
                        )

show = df[['Name', 'Fare', 'Fare_Category']].head(15)

print(show)
```

```python
#7
"""
Create a separate DataFrame containing the average Fare and average Age for each 
Sex. 
Merge it back into the Titanic DataFrame using Sex. 
Display the first 10 passengers with the new information.
"""

summary = df.groupby('Sex').agg(
    Average_Fare=('Fare', 'mean'),
    Average_Age=('Age', 'mean')
).reset_index()

result = df.merge(summary, on='Sex')

show = result.head(10)

print(show)
```

```python
#8
"""
Create: 
FamilySize = SibSp + Parch + 1 
FarePerPerson = Fare / FamilySize 

Then create a NumPy matrix containing: 
Pclass, Age, FamilySize, Fare, FarePerPerson 

Replace missing Age with the median and print the matrix shape. 
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['FarePerPerson'] = df['Fare'] / df['FamilySize']

pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
familysize = df['FamilySize'].to_numpy()
fare = df['Fare'].to_numpy()
fareperperson = df['FarePerPerson'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

mat = np.column_stack((pclass,age,familysize,fare,fareperperson))

print(mat.shape)
```

```python
#9 
"""
Create Fare_ZScore using: 
(Fare - mean Fare) / standard deviation of Fare 

Display the 5 passengers with the largest absolute Z-score.
"""

df['Fare_ZScore'] = ((df['Fare'] - df['Fare'].mean())/ df['Fare'].std())

df['Abs_ZScore'] = df['Fare_ZScore'].abs()

show = df.sort_values('Abs_ZScore',ascending=False).head(5)

print(show)
```

```python
#10
"""
Create these features: 
FamilySize, FarePerPerson, PclassAvgFare, FarePercentOfClassAvg 

Then create a NumPy matrix containing: 
Survived, Pclass, Age, Fare, FamilySize, FarePerPerson 

Replace missing Age with the median, standardize FarePerPerson, transpose the 
matrix, and print its shape.
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

df['FarePerPerson'] = df['Fare'] / df['FamilySize']

df['PclassAvgFare'] = df.groupby('Pclass')['Fare'].transform('mean')

df['FarePercentOfClassAvg'] = (df['Fare'] / df['PclassAvgFare']) * 100

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
fare = df['Fare'].to_numpy()
familysize = df['FamilySize'].to_numpy()
fareperperson = df['FarePerPerson'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

fareperperson = (fareperperson - fareperperson.mean()) / fareperperson.std()

mat = np.column_stack((survived, pclass, age, fare, familysize, fareperperson))

trans = mat.T

print(trans.shape)
```
