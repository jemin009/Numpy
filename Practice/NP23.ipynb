```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1
"""
Using Pclass, calculate the average Fare for each class. 
Then create a new column: 
Fare_Difference 

It should show how far each passenger's Fare is from their class average. 

Finally, display the 10 passengers with the largest absolute difference.
"""

df['Average_Fare'] = df.groupby('Pclass')['Fare'].transform('mean')

df['Fare_Difference'] = (df['Fare'] - df['Average_Fare']).abs()

show = df.sort_values('Fare_Difference', ascending=False).head(10)

print(show)
```

```python
#2
"""
Create a NumPy-based Age_Group column using these rules: 
 Age < 18 → "Child"  
 Age 18–59 → "Adult"  
 Age ≥ 60 → "Senior"  

Replace missing Age values with the median Age before creating the groups. 

Display: 
Name, Age, Age_Group, Survived 
for the first 15 passengers. 
"""

age = df['Age'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

df['Age_Group'] = np.where(age < 18, 'Child', np.where((age >= 18) & (age <= 59),'Adult', 'Senior'))

show = df[['Name','Age','Age_Group','Survived']].head(15)

print(show)
```

```python
#3
"""
Create a pivot table showing the median Fare for each combination of: 
 Pclass  
 Sex  

The result should have: 
 Pclass as rows  
 Sex as columns  
 median Fare as values 
"""

pvt = df.pivot_table(
                        index = 'Pclass',
                        columns = 'Sex',
                        values = 'Fare',
                        aggfunc = 'median'
).reset_index()

print(pvt)
```

```python
#4
"""
Create a NumPy array from Fare and classify passengers: 
 Fare < 20 → "Low"  
 Fare 20–100 → "Medium"  
 Fare > 100 → "High"  

Create: 
Fare_Category 
Then find the number of passengers in each category.
"""

fare = df['Fare'].to_numpy()

df['Fare_Category'] = np.where(fare < 20, 'Low',
                                np.where((fare >= 20) & (fare <= 100), 'Medium','High'))

grp = df['Fare_Category'].value_counts()

print(grp)
```

```python
#5
"""
Create: 
FarePerPerson = Fare / FamilySize 

where: 
FamilySize = SibSp + Parch + 1 
Then find the 10 passengers with the highest FarePerPerson. 

Display: 
Name, Pclass, FamilySize, Fare, FarePerPerson
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

df['FarePerPerson'] = df['Fare'] / df['FamilySize']

sort = df.sort_values('FarePerPerson', ascending = False).head(10)

show = sort[['Name','Pclass','FamilySize','Fare','FarePerPerson']]

print(show)
```

```python
#6
"""
Create a pivot table showing survival rate (%) for each: 
Pclass × Sex 

Then identify which Pclass + Sex combination has the highest survival rate. 
"""

pvt = df.pivot_table(
                        index = 'Sex',
                        columns = 'Pclass',
                        values = 'Survived',
                        aggfunc = 'mean'
) * 100

print(pvt)
print()

print("Highest survival rate:", pvt.max().max())
print("Combination:", pvt.stack().idxmax())
```

```python
#7
"""
Convert Fare to a NumPy array. 

Perform these operations: 
1. Replace any negative values with 0 using NumPy.  
2. Add 10% to every fare.  
3. Calculate the new mean fare.  
4. Print the mean.
"""

fare = df['Fare'].to_numpy()

fare = np.where(fare < 0, 0, fare)

fare = fare + (fare * 0.10)

print(fare.mean())
```

```python
#8
"""
Calculate the average Age for each Pclass. 

Create: 
Age_Difference 
as the absolute difference between each passenger's Age and their Pclass average Age. 

Then display the 10 passengers whose Age is farthest from their class average. 
"""

df['AvgAge'] = df.groupby('Pclass')['Age'].transform('mean')

df['Age_Difference'] = (df['Age'] - df['AvgAge']).abs()

sort = df.sort_values('Age_Difference', ascending = False).head(10)

print(sort)
```

```python
#9
"""
Create these columns: 
FamilySize 
FarePerPerson 

Then create a NumPy matrix containing: 
Pclass 
Age 
FamilySize 
Fare 
FarePerPerson 
Survived 

Requirements: 
 Replace missing Age with median Age.  
 Min-max normalize Fare.  
 Standardize FarePerPerson.  
 Print the matrix shape. 
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

df['FarePerPerson'] = df['Fare'] / df['FamilySize']

pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
familysize = df['FamilySize'].to_numpy()
fare = df['Fare'].to_numpy()
fareperperson = df['FarePerPerson'].to_numpy()
survived = df['Survived'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)
fare = (fare - fare.min()) / (fare.max() - fare.min())

fareperperson = (fareperperson - fareperperson.mean()) / fareperperson.std()


mat = np.column_stack((pclass, age, familysize, fare, fareperperson, survived))

print(mat.shape)
```

```python
#10
"""
Create these features: 
FamilySize 
FarePerPerson 
PclassAvgFare 
FarePercentOfClassAvg 

Then: 
1. Create a NumPy matrix containing Survived, Pclass, Age, FarePerPerson, and FarePercentOfClassAvg.  
2. Replace missing Age with median Age.  
3. Find the 5 passengers with the lowest FarePercentOfClassAvg among passengers whose percentage is below 100%.  

Display: 
Name, Pclass, Fare, PclassAvgFare, FarePercentOfClassAvg 
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

df['FarePerPerson'] = df['Fare'] / df['FamilySize']

df['PclassAvgFare'] = df.groupby('Pclass')['Fare'].transform('mean')

df['FarePercentOfClassAvg'] = (df['Fare'] / df['PclassAvgFare']) * 100

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
fareperperson = df['FarePerPerson'].to_numpy()
farepercentofclassavg = df['FarePercentOfClassAvg'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

mat = np.column_stack((survived,pclass,age,fareperperson,farepercentofclassavg))

print("Matrix shape:", mat.shape)


sort = df[df['FarePercentOfClassAvg'] < 100].sort_values('FarePercentOfClassAvg',ascending=True).head(5)

show = sort[['Name','Pclass','Fare','PclassAvgFare','FarePercentOfClassAvg']]

print(show)
```
