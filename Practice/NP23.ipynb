```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1
"""
Using Pclass, calculate the average Fare for each class. 
Then create a new column: 
Fare_Difference 

It should show how far each passenger's Fare is from their class average. 

Finally, display the 10 passengers with the largest absolute difference.
"""

df['Average_Fare'] = df.groupby('Pclass')['Fare'].transform('mean')

df['Fare_Difference'] = (df['Fare'] - df['Average_Fare']).abs()

show = df.sort_values('Fare_Difference', ascending=False).head(10)

print(show)
```

```python
#2
"""
Create a NumPy-based Age_Group column using these rules: 
 Age < 18 → "Child"  
 Age 18–59 → "Adult"  
 Age ≥ 60 → "Senior"  

Replace missing Age values with the median Age before creating the groups. 

Display: 
Name, Age, Age_Group, Survived 
for the first 15 passengers. 
"""

age = df['Age'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

df['Age_Group'] = np.where(age < 18, 'Child', np.where((age >= 18) & (age <= 59),'Adult', 'Senior'))

show = df[['Name','Age','Age_Group','Survived']].head(15)

print(show)
```

```python
#3
"""
Create a pivot table showing the median Fare for each combination of: 
 Pclass  
 Sex  

The result should have: 
 Pclass as rows  
 Sex as columns  
 median Fare as values 
"""

pvt = df.pivot_table(
                        index = 'Pclass',
                        columns = 'Sex',
                        values = 'Fare',
                        aggfunc = 'median'
).reset_index()

print(pvt)
```
