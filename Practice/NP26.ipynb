```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1
"""
1. Convert Fare into a NumPy array without modifying the original column. 
2. Replace missing fares with the median using np.where(). 
3. Calculate the mean and standard deviation of the fares. 
4. Select passengers whose fare is greater than the mean using Boolean masking. 
5. Display the number of selected passengers and the 10 highest fares, showing Name, Pclass, and Fare. 
"""

fare = df['Fare'].to_numpy()

fare = np.where(np.isnan(fare), np.nanmedian(fare), fare)

print(f"Mean of fare : {fare.mean():.2f}")
print(f"standard deviation of fare : {fare.std():.2f}")

fare_mask = (fare > fare.mean())

print("Number of selected passengers:", fare_mask.sum())

show = df.loc[fare_mask,['Name','Fare','Pclass']].sort_values('Fare', ascending = False).head(10)

print(show)
```

```python
#2
"""
1. Fill missing Age values with the median in a NumPy array. 
2. Calculate the mean and standard deviation of age. 
3. Calculate the z-score for each passenger's age. 
4. Select passengers with a z-score greater than 1. 
5. Display the 10 highest z-scores with Name, Age, and the z-score.
"""

age = df['Age'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

mean_age = age.mean()
std_age = age.std()

print(f"Mean of age : {age.mean():.2f}")
print(f"Standard deviation of age : {age.std():.2f}")

z_score = (age - mean_age) / std_age
z_score_mask = z_score > 1

print("Selected passengers:", z_score_mask.sum())

show = df.loc[z_score_mask, ['Name', 'Age']].copy()

show['ZScore'] = z_score[z_score_mask]

show = show.sort_values('ZScore', ascending=False).head(10)

print(show)
```

```python
#3
"""
1. Create FamilySize = SibSp + Parch + 1. 
2. Create FamilyCategory: Alone for size 1, Small for size 2–4, and Large for size greater than 4. 
3. Calculate the survival rate for each category. 
4. Calculate the passenger count for each category. 
5. Display the summary sorted by survival rate, highest first.
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

df['FamilyCategory'] = np.where(df['FamilySize'] == 1, 'Alone',
                            np.where(df['FamilySize'] <= 4, 'Small', 'Large')
                        )

summary  = df.groupby('FamilyCategory').agg(
                        Survival_rate = ('Survived','mean'),
                        Passenger_count = ('Survived','count')
)

summary = summary.sort_values('Survival_rate', ascending=False)

print(summary)
```
