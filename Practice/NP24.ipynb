```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1
"""
Create a feature called PclassMedianFare using the median Fare for each passenger class. 

Then create FareVsMedian, the absolute difference between each passenger's Fare and their class's median Fare. 

Display the 10 passengers with the largest differences, showing Name, Pclass, Fare, PclassMedianFare, and FareVsMedian.
"""

df['PclassMedianFare'] = df.groupby('Pclass')['Fare'].transform('median')

df['FareVsMedian'] = (df['Fare'] - df['PclassMedianFare']).abs()

sort = df.sort_values('FareVsMedian', ascending = False).head(10)

show = sort[['Name','Pclass','Fare','PclassMedianFare','FareVsMedian']]

print(show)
```

```python
#2
"""
Use NumPy to create an AgeGroup feature: 
 Child: Age below 18 
 Adult: Age from 18 through 59 
 Senior: Age 60 or above 
 
Replace missing Age values with the median Age before assigning groups. 
Calculate the survival rate for each age group as a percentage. 
Display the results from highest to lowest survival rate.
"""

age = df['Age'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

df['AgeGroup'] = np.where(age < 18, 'Child',
                            np.where((age >= 18) & (age <= 59),'Adult','Senior'))

survival_rate = (df.groupby('Survived')['Age'].transform('mean') * 100).sort_values(ascending = False)

print(survival_rate)
```

```python
#3
"""
Create a pivot table showing the average FarePerPerson for each combina on of 
Pclass and Sex. 
 
First create: 
FamilySize = SibSp + Parch + 1 
FarePerPerson = Fare / FamilySize 
 
Use Pclass as rows and Sex as columns.
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['FarePerPerson'] = df['Fare'] / df['FamilySize']

pvt = df.pivot_table(
                        index = 'Pclass',
                        columns = 'Sex',
                        values = 'FarePerPerson',
                        aggfunc = 'mean'
) 

print(pvt)
```

```python
#4
"""
Use NumPy to classify passengers into three fare categories: 
 Budget: Fare below 15 
 Standard: Fare from 15 through 75 
 Premium: Fare above 75 
 
Create FareCategory, then display the passenger count and percentage of the dataset 
in each category. 
"""

fare = df['Fare'].to_numpy()

df['FareCategory'] = np.where(fare < 15, 'Budget',
                            np.where((fare >= 15) & (fare <= 75), 'Standard', 'Premium'))

counts = df['FareCategory'].value_counts()
percentages = df['FareCategory'].value_counts(normalize=True) * 100

result = pd.DataFrame({'PassengerCount': counts, 'Percentage': percentages})

print(result)
```

```python
#5
"""
Calculate each passenger's class-average fare using transform('mean'). 
 
Create FareRatio, calculated as: 
Fare / PclassAvgFare 
 
Find the 10 passengers with the highest FareRatio. Display their names, passenger 
class, fare, class-average fare, and FareRatio.
"""

df['PclassAvgFare'] = df.groupby('Pclass')['Fare'].transform('mean')

df['FareRatio'] = df['Fare'] / df['PclassAvgFare']

show = df[['Name','Pclass','Fare','PclassAvgFare','FareRatio']].sort_values('FareRatio', ascending = False).head(10)

print(show)
```

```python
#6
"""
Create a pivot table showing the survival rate as a percentage for each Pclass and Sex combination. 

Then identify the combination with the highest survival rate. 
 
Hint: stack().idxmax() can help you idetify the row and column combination. 
"""

pvt = df.pivot_table(
                        index = 'Pclass',
                        columns = 'Sex',
                        values = 'Survived',
                        aggfunc = 'mean'
)* 100

show = pvt.stack().idxmax()

print(pvt)
print(show)
```

```python
#7
"""
Convert the Fare column into a NumPy array. 
1. Replace missing values with the median Fare. 
2. Replace any negtive values with zero. 
3. Increase every fare by 8%. 
4. Print the minimum, maximum, and mean of the transformed array. 
 
Do not modify the original df['Fare'] column.
"""

fare = df['Fare'].to_numpy().copy()

fare = np.where(np.isnan(fare), np.nanmedian(fare), fare)

fare = np.where(fare < 0, 0, fare)

fare = fare + (fare * 0.08)

print(f"Minimum : {fare.min()}")
print(f"Maximum : {fare.max()}")
print(f"Mean : {fare.mean()}")

print(fare)
```

```python
#8
"""
Calculate the average Age for each Pclass using transform('mean'). 

Replace missing Age values with the overall median Age for this analysis. 

Create AgeVsClassAvg, the absolute difference between each passenger's Age and their class's average Age. 

Display the 10 passengers whose ages differ most from their class average, showing 
Name, Pclass, Age, class-average Age, and AgeVsClassAvg.
"""

df['Age'] = np.where(np.isnan(df['Age']), np.nanmedian(df['Age']), df['Age'])

df['AvgAgePclass'] = df.groupby('Pclass')['Age'].transform('mean')

df['AgeVsClassAvg'] = (df['Age'] - df['AvgAgePclass']).abs()

show = df[['Name','Pclass','Age','AvgAgePclass','AgeVsClassAvg']].sort_values('AgeVsClassAvg', ascending = False).head(10)

print(show)
```

```python
#9
"""
Create these features: 
 FamilySize 
 FarePerPerson 
 PclassAvgFare 
 FarePercentOfClassAvg 
 
Create a NumPy matrix containing these six columns in this order: 
1. Survived 
2. Pclass 
3. Age 
4. FamilySize 
5. FarePerPerson 
6. FarePercentOfClassAvg 
 
Replace missing Age values with the median Age in the array used for the matrix. 
Finally, print the matrix shape and the mean of the FarePercentOfClassAvg column.
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['FarePerperson'] = df['Fare'] / df['FamilySize']
df['PclassAvgFare'] = df.groupby('Pclass')['Fare'].transform('mean')
df['FarePercentOfClassAvg'] = (df['Fare'] / df['PclassAvgFare']) * 100

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
familysize = df['FamilySize'].to_numpy()
fareperperson = df['FarePerPerson'].to_numpy()
farepercentofclassavg = df['FarePercentOfClassAvg'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

mat = np.column_stack((survived, pclass, age, familysize, fareperperson, farepercentofclassavg))

print(f"Mean of FarePercentOfClassAvg : {mat[:,:5].mean()}")

print(mat.shape)
```

```python
#10
"""
Create these features: 
 FamilySize 
 FarePerPerson 
 PclassAvgFare 
 FarePercentOfClassAvg 
 
Then: 
1. Replace missing Age values with the median Age in a NumPy array. 
2. Create a NumPy matrix containing Survived, Pclass, Age, FarePerPerson, and FarePercentOfClassAvg. 
3. Select passengers whose FarePercentOfClassAvg is below 100%. 
4. From those passengers, find the five with the largest FamilySize. 
 
Display Name, Survived, Pclass, Age, FamilySize, FarePerPerson, and FarePercentOfClassAvg.
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['FarePerperson'] = df['Fare'] / df['FamilySize']
df['PclassAvgFare'] = df.groupby('Pclass')['Fare'].transform('mean')
df['FarePercentOfClassAvg'] = (df['Fare'] / df['PclassAvgFare']) * 100

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
fareperperson = df['FarePerPerson'].to_numpy()
farepercentofclassavg = df['FarePercentOfClassAvg'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

mat = np.column_stack((survived, pclass, age, fareperperson, farepercentofclassavg))

show = (
    df.loc[df['FarePercentOfClassAvg'] < 100,
           ['Name', 'Survived', 'Pclass', 'Age', 'FamilySize','FarePerPerson', 'FarePercentOfClassAvg']]
    .sort_values('FamilySize', ascending=False)
    .head(5)
)

print(show)
```
