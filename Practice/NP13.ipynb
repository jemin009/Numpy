```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1 Using the Fare NumPy array, find the 3 highest unique fare values. 

fare = df['Fare'].to_numpy()
unique_fare = np.unique(fare)

print(unique_fare[-3:][::-1])
```

```python
#2 Using Age, find the index positions of all passengers whose age is between 20 and 30, using NumPy vectorized operations. 

age = df['Age'].to_numpy()

print(np.where((age >= 20) & (age <= 30)))
```

```python
#3 Using Pclass and Survived, create a NumPy matrix containing both columns and display its shape. 

pclass = df['Pclass'].to_numpy()
survived = df['Survived'].to_numpy()

mat = np.column_stack((pclass,survived))

mat
```

```python
#4 Using SibSp and Parch, create a NumPy matrix and calculate the total family connections for each passenger using vectorized NumPy operations. 

sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()

family_connections = sibsp + parch
print(family_connections)
```

```python
#5 Using Fare, find the indices of the 10 passengers with the lowest non-zero fares using argsort(). 

fare = df['Fare'].to_numpy()

indices = np.where(fare > 0)[0]

sorted_indices = indices[np.argsort(fare[indices])]

print(sorted_indices[:10])
```

```python
#6 Using the Pclass, SibSp, and Parch arrays, create a feature matrix and transpose it for feature-wise analysis.

pclass = df['Pclass'].to_numpy() 
sibsp = df['SibSp'].to_numpy() 
parch = df['Parch'].to_numpy() 

mat = np.column_stack((pclass,sibsp,parch))

mat.transpose() 
```

```python
#7 Using the Fare array, increase every fare by 20% for passengers who paid more than 100, while leaving all other fares unchanged. Use NumPy vectorized operations without a loop. 

fare = df['Fare'].to_numpy()

increase = np.where(fare > 100, fare + (fare * 20) / 100, fare)

print(increase)
```

```python
#8 Using the Age array, calculate the mean age separately for passengers in each Pclass using NumPy operations. 

age = df['Age'].to_numpy()
pclass = df['Pclass'].to_numpy()

print(np.nanmean(age[pclass == 1]))
print(np.nanmean(age[pclass == 2]))
print(np.nanmean(age[pclass == 3]))
```

```python
#9 
"""
Create a NumPy matrix containing: 
Survived, Pclass, Age, SibSp, Parch, Fare 

Then split the matrix into two row-wise parts using NumPy.
"""

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()

mat = np.column_stack((survived,pclass, age, sibsp, parch, fare))

print(np.array_split(mat,2))
```

```python
#10 Using the Fare array, calculate the z-score standardized Fare values using NumPy vectorized operations, then verify that the standardized array has a mean approximately equal to 0.

mean = np.mean(fare)
std = np.std(fare)

z_score = (fare - mean) / std

print(z_score)
print(np.mean(z_score))
```
