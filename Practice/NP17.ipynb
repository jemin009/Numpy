```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1 Using Pandas, find the average Fare for each Pclass using groupby() and agg(). 

df.groupby('Pclass')['Fare'].agg('mean')

df
```

```python
#2 Using Pandas string operations, find all passengers whose Name contains "William". Display only Name, Sex, Age, and Fare.

will = df.loc[df['Name'].str.contains('William'),['Name','Sex','Age','Fare']]

will
```

```python
#3 
"""
Using Pandas, create a new column called FamilySize: 
FamilySize = SibSp + Parch + 1 

Then display Name, FamilySize, and Survived for the 10 passengers with the largest 
FamilySize. 
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

large = df.sort_values('FamilySize', ascending=False)

large[['Name', 'FamilySize', 'Survived']].head(10)
```

```python
#4 
"""
Using groupby() and agg(), find for each Sex: 
 average Age  
 average Fare  
 survival rate 
"""

sex = df.groupby('Sex').agg({
    'Age': 'mean',
    'Fare': 'mean',
    'Survived': 'mean'
})

sex['Survived'] = sex['Survived'] * 100

sex
```

```python
#5 Convert the Fare column to a NumPy array and find the indices of all passengers whose Fare is greater than the overall average Fare. Use those indices to display their Name and Fare.

fare = df['Fare'].to_numpy()

average = np.mean(fare)

ind = np.where(fare > average)[0]

result = df.loc[ind, ['Name', 'Fare']]

print(result)
```

```python
#6 
"""
Using Pandas, create a new column FarePerFamilyMember: 
Fare / FamilySize 

Handle any possible division-by-zero issue appropriately. Display the 10 passengers 
with the highest FarePerFamilyMember. 
"""

df['FarePerFamilyMember'] = df['Fare'] / df['FamilySize']

sort = df.sort_values('FarePerFamilyMember', ascending=False)

print(sort[['Name', 'FamilySize', 'Fare', 'FarePerFamilyMember']].head(10))
```

```python
#7 
"""
Create a NumPy feature matrix containing: 
Pclass, Age, SibSp, Parch, Fare, FamilySize 

Replace missing Age values with the median Age, then display: 
 matrix shape  
 first 5 rows  
"""

pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()
familysize = df['FamilySize'].to_numpy()

median = np.nanmedian(age)

age = np.where(np.isnan(age),median,age)

mat = np.column_stack((pclass,age,sibsp,parch,fare,familysize))

print(mat.shape)

print(mat[:5])
```

```python
#8 
"""
Using Pandas groupby() and agg(), calculate for each Pclass: 
 minimum Fare  
 maximum Fare  
 average Fare  
 median Age  
"""

grp = df.groupby('Pclass').agg(
    MinFare=('Fare', 'min'),
    MaxFare=('Fare', 'max'),
    AverageFare=('Fare', 'mean'),
    MedianAge=('Age', 'median')
)

print(grp)
```

```python
#9 
"""
Using NumPy, standardize the Fare column: 
z = (x - mean) / standard deviation 

Create a new DataFrame column called Fare_ZScore, then display the 5 passengers 
with the highest absolute z-score.
"""

df['Fare_ZScore'] = ((df['Fare'] - df['Fare'].mean()) / df['Fare'].std())

result = df.loc[
    df['Fare_ZScore'].abs().sort_values(ascending=False).head(5).index,
    ['Name', 'Age', 'Fare_ZScore']
]

print(result)
```

```python
#10 
"""
Create a final feature matrix containing: 
Survived, Pclass, Age, SibSp, Parch, Fare, FamilySize 

Then: 
 Replace missing Age with the median Age.  
 Create FamilySize.  
 Min-max normalize Fare.  
 Convert the final DataFrame to a NumPy matrix.  
 Transpose the matrix.  
 Print the final matrix shape.
"""

survived = df['Survived'].to_numpy()
pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()
familysize = df['FamilySize'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

fare = (fare - np.min(fare)) / (np.max(fare) - np.min(fare))

mat = np.column_stack((survived, pclass, age, sibsp, parch, fare, familysize))

final = mat.T

print(final.shape)
```
