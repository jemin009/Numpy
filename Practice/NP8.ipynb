```python
import numpy as np
```

```python
#1
"""
A company has monthly sales targets from ₹25,000 to ₹75,000, increasing by ₹5,000. 
 
Using NumPy: 
1. Create the target array using np.arange().  
2. Find the total target.  
3. Find the average target.  
4. Find the highest target.  
5. Extract the first 4 targets.  
6. Extract the last 3 targets.  
7. Increase all targets by 12%.  
8. Find the new average target.  
9. Find the new maximum target.  
10. Display shape and ndim. 
"""

sales = np.arange(25000,75001,5000)

print(f"Total target : {sales.sum()}")
print(f"Average target : {sales.mean()}")
print(f"Highest target : {sales.max()}")
print(f"first 4 targets : {sales[:4]}")
print(f"last 3 targets : {sales[-3:]}")

increased = sales + (sales * 12) / 100

print(f"new average target : {increased.mean()}")
print(f"new maximum target : {increased.max()}")
print(f"Shape : {increased.shape}")
print(f"Ndim : {increased.ndim}")
```

```python
#2
"""
HR wants 8 evenly spaced performance scores from 55 to 90, including both 
endpoints. 
 
Using np.linspace(): 
1. Create the array.  
2. Find the average score.  
3. Find the maximum and minimum.  
4. Extract the first 3 scores.  
5. Extract the last 2 scores.  
6. Add 5 points to every score.  
7. Find the new maximum.  
8. Display shape and ndim. 
"""

scores = np.linspace(55,90,8)

print(f"Array : {scores}")
print(f"average score : {scores.mean()}")
print(f"maximum score : {scores.max()}")
print(f"minimum score : {scores.min()}")
print(f"first 3 scores : {scores[:3]}")
print(f"last 2 scores : {scores[-2:]}")

add = scores + 5

print(f"new maximum : {add.max()}")
print(f"shape : {add.shape}")
print(f"ndim : {add.ndim}")
```

```python
#3
"""
Create a NumPy array and: 
1. Display its shape and ndim.  
2. Find total sales for each store.  
3. Find average sales for each store.  
4. Find total sales for each quarter.  
5. Extract Q2 and Q3 for all stores.  
6. Extract all quarters for Store 4.  
7. Find the highest individual sales value.  
8. Find the lowest individual sales value.  
9. Increase only Q4 sales by 10%.  
10. Find the new highest individual sales value. 
"""

sales = [ 
    [12000, 15000, 18000, 21000], 
    [14000, 16000, 19000, 23000], 
    [11000, 17000, 20000, 22000], 
    [15000, 18000, 21000, 25000], 
    [13000, 15500, 17500, 24000] 
]

arr = np.array(sales)

print(f"Array : {arr}")
print(f"Shape : {arr.shape}")
print(f"Ndim : {arr.ndim}")
print(f"total sales for each store : {arr.sum(axis=1)}")
print(f"average sales for each store : {arr.mean(axis=1)}")
print(f"total sales for each quarter : {arr.sum(axis=0)}")
print(f"Q2 and Q3 for all stores : {arr[:,1:3]}")
print(f"all quarters for Store 4 : {arr[3]}")
print(f"highest individual sales value : {arr.max(axis=1)}")
print(f"lowest individual sales value : {arr.min(axis=1)}")

arr[:,-1] = arr[:,-1] + (arr[:,-1] * 10) / 100

print(f"new highest individual sales value : {arr.max(axis=1)}")
```

```python
#6
"""
Generate 15 random values using a NumPy random function you have learned. 
Treat them as model accuracy values. 
 
Tasks: 
1. Generate 15 random values.  
2. Find the average accuracy.  
3. Find the maximum accuracy.  
4. Find the minimum accuracy.  
5. Extract the first 5 values.  
6. Add 0.10 to every value.  
7. Find the new maximum.  
8. Find the new average.  
9. Check whether the new maximum is greater than 1.  
10. Display shape and ndim.  
"""

num = np.random.random(15)

print(f"Values : {num}")
print(f"average accuracy : {num.mean()}")
print(f"maximum accuracy : {num.max()}")
print(f"minimum accuracy : {num.min()}")
print(f"first 5 values : {num[:5]}")

new = num + 0.10

print(f"the new maximum : {new.max()}")
print(f"the new average : {new.mean()}")

if new.max() > 1:
    print("Yes, the new maximum is greater than 1")
else:
    print("No, the new maximum is not greater than 1")

print(f"Shape : {new.shape}")
print(f"Ndim : {new.ndim}")
```

```python
#7
"""
Create these two matrices: 
A = np.eye(5) 
B = np.identity(5) 
 
Tasks: 
1. Display both matrices.  
2. Display their shapes.  
3. Display their dimensions.  
4. Find the maximum value of each.  
5. Find the minimum value of each.  
6. Calculate the total of each matrix.  
7. Compare whether both matrices have the same values.  
8. Extract the third row from both matrices.  
9. Extract the third column from both matrices.  
10. Explain in one sentence why np.eye(5) and np.identity(5) produce the same 
5×5 result here. 
"""

A = np.eye(5)
B = np.identity(5)

print(f"A : {A}")
print(f"B : {B}")

print(f"Shape of A : {A.shape}")
print(f"Shape of B : {B.shape}")

print(f"Dimensions of A : {A.ndim}")
print(f"Dimensions of B : {B.ndim}")

print(f"maximum value of A : {A.max()}")
print(f"maximum value of B : {B.max()}")

print(f"minimum value of A : {A.min()}")
print(f"minimum value of B : {B.min()}")

print(f"total of A : {A.sum()}")
print(f"total of B : {B.sum()}")

if np.array_equal(A,B):
    print("Yes, both matrices have the same values")
else:
    print("No, both matrices haven't the same values")

print(f"third row from A : {A[2]}")
print(f"third row from B : {B[2]}")

print(f"third column from A : {A[:,2]}")
print(f"third column from B : {B[:,2]}")

print("Both produce the same 5x5 result because we put a same input in both but using eye we can create flexible matrix but can't with identity it can produce in square matrix")
```
