```python
import numpy as np
```

```python
#1
"""
Using NumPy: 
1. Create the array.  
2. Find total sales.  
3. Find average sales.  
4. Find the highest and lowest sales.  
5. Extract sales of the first 4 days.  
6. Extract sales of the last 3 days.  
7. Check whether the highest sales are greater than 20000.  
8. Print shape and ndim.  
"""

sales = [12500, 14800, 13200, 17600, 15900, 19200, 18500, 21000] 

arr = np.array(sales)

print(f"Total sales : {arr.sum()}")
print(f"Average sales : {arr.mean()}")
print(f"Highest sales : {arr.max()}")
print(f"Lowest sales : {arr.min()}")
print(f"Sales of the first 4 days : {arr[:4]}")
print(f"Sales of the last 3 days : {arr[-3:]}")

if arr.max() > 20000:
    print("Yes, the highest sales are greater than 20000")
else:
    print("No, the highest sales are not greater than 20000")

print(f"Shape : {arr.shape}")
print(f"Ndim : {arr.ndim}")
```

```python
#2
"""
Using NumPy: 
1. Create the array.  
2. Find average rating.  
3. Find highest and lowest rating.  
4. Extract ratings from the 3rd to 7th customer.  
5. Extract the last 3 ratings.  
6. Find the rating of the 5th customer.  
7. Check whether average rating is at least 4.0.  
8. Print dtype.  
"""

ratings = [4.1, 3.8, 4.5, 4.7, 3.9, 4.8, 4.2, 4.6, 3.7] 

arr = np.array(ratings)

print(f"Average rating : {arr.mean()}")
print(f"Highest rating : {arr.max()}")
print(f"Lowest rating : {arr.min()}")
print(f"ratings from the 3rd to 7th customer : {arr[2:7]}")
print(f"the last 3 ratings : {arr[-3:]}")
print(f"the rating of the 5th customer : {arr[4]}")

if arr.mean() >= 4.0:
    print(f"Yes, average rating is at least 4.0")
else:
    print(f"No, average rating is not at least 4.0")

print(f"Dtype : {arr.dtype}")
```

```python
#3
"""
Using NumPy: 
1. Create the 2D array.  
2. Print shape and ndim.  
3. Find the salary of Employee 3 in February.  
4. Extract the complete salary record of Employee 2.  
5. Extract January and February salaries of all employees.  
6. Find the highest salary.  
7. Find the lowest salary.  
8. Find the total salary of Employee 4.  

Rows = Employees 
Columns = Jan, Feb, Mar
"""

salary = [ 
    [28000, 30000, 32000], 
    [35000, 37000, 39000], 
    [42000, 44000, 46000], 
    [31000, 33000, 35000] 
] 

arr = np.array(salary)

print(f"Shape : {arr.shape}")
print(f"Ndim : {arr.ndim}")
print(f"the salary of Employee 3 in February : {arr[2][1]}")
print(f"the complete salary record of Employee 2 : {arr[1]}")
print(f"January and February salaries of all employees : {arr[:,:2]}")
print(f"January and February salaries of all employees : {arr[:,:2]}")
print(f"Highest salary : {arr.max()}")
print(f"Lowest salary : {arr.min()}")
print(f"total salary of Employee 4 : {arr.sum(axis=1)[-1]}")
```

```python
#4
"""
Using NumPy: 
1. Create the 2D array.  
2. Find total annual sales of North.  
3. Find total annual sales of West.  
4. Extract Q3 and Q4 sales of all regions.  
5. Find the Q2 sales of South.  
6. Find the highest quarterly sales.  
7. Find the lowest quarterly sales.  
8. Print shape and ndim. 

Rows = North, South, West 
Columns = Q1, Q2, Q3, Q4
"""

sales = [ 
    [12000, 15000, 18000, 21000], 
    [14000, 16000, 19000, 23000], 
    [11000, 17000, 20000, 25000] 
]

arr = np.array(sales)

print(f"total annual sales of North : {arr.sum(axis=1)[0]}")
print(f"total annual sales of West : {arr.sum(axis=1)[-1]}")
print(f"Q3 and Q4 sales of all regions : {arr[:,-2:]}")
print(f"Q2 sales of South : {arr[1][1]}")
print(f"highest quarterly sales : {arr.max()}")
print(f"lowest quarterly sales : {arr.min()}")
print(f"shape : {arr.shape}")
print(f"ndim : {arr.ndim}")
```

```python
#5
"""
Using NumPy: 
1. Create the array.  
2. Increase every price by ₹150.  
3. Find the average of the new prices.  
4. Find the highest new price.  
5. Extract the first 5 new prices.  
6. Extract the last 2 new prices.  
7. Check whether the highest new price is greater than ₹2000. 
"""

prices = [499, 799, 999, 1299, 1499, 1799, 1999] 

arr = np.array(prices)

new = arr + 150

print(f"average of the new prices : {new.mean()}")
print(f"highest new price : {new.max()}")
print(f"first 5 new prices : {new[:5]}")
print(f"last 2 new prices : {new[-2:]}")

if new.max() > 2000:
    print("Yes, the highest new price is greater than ₹2000")
else:
    print("No, the highest new price is not greater than ₹2000")
```

```python
#6
"""
Using NumPy: 
1. Create the array.  
2. Increase every temperature by 2°C.  
3. Find the average temperature after the increase.  
4. Find the highest temperature after the increase.  
5. Extract values from index 3 to 8.  
6. Extract the last 4 values.  
7. Check whether any recorded temperature after increase is greater than 38°C. 
"""

temperature = [27, 29, 31, 30, 33, 35, 34, 37, 36, 32, 31, 30] 

arr = np.array(temperature)

new = arr + 2

print(new)
print(f"average temperature after the increase : {new.mean()}")
print(f"highest temperature after the increase : {new.max()}")
print(f"values from index 3 to 8 : {new[3:9]}")
print(f"last 4 values : {new[-4:]}")

if (new > 38).any():
    print("Yes, recorded temperature after increase is greater than 38°C")
else:
    print("No, recorded temperature after increase is greater than 38°C")
```

```python
#8
"""
Create the following arrays: 
1. 4 × 3 zeros.  
2. 3 × 4 ones.  
3. 2 × 5 array filled with 50.  
4. 3 × 3 array filled with 999.  
5. 2 × 4 empty array.  
Then print the shape and ndim of every array.
"""

zeros = np.zeros([4,3])
print("Zeros :-")
print(zeros)
print(f"Shape : {zeros.shape}")
print(f"Ndim : {zeros.ndim}")
print()

ones = np.ones([3,4])
print("Ones :-")
print(ones)
print(f"Shape : {ones.shape}")
print(f"Ndim : {ones.ndim}")
print()

Full = np.full((2,5),50)
print("Full :-")
print(Full)
print(f"Shape : {Full.shape}")
print(f"Ndim : {Full.ndim}")
print()

full = np.full((3,3),999)
print("Full :-")
print(full)
print(f"Shape : {full.shape}")
print(f"Ndim : {full.ndim}")
print()

empty = np.empty([2,4])
print("Emptry :-")
print(empty)
print(f"Shape : {empty.shape}")
print(f"Ndim : {empty.ndim}")
print()
```

```python
#10
"""
Using NumPy: 
1. Create the 2D array.  
2. Print shape, ndim, and dtype.  
3. Find annual sales of each product.  
4. Determine which product has the highest annual sales.  
5. Find Q4 sales of Product B.  
6. Extract Q1 and Q2 sales of all products.  
7. Find the highest quarterly sales.  
8. Increase all sales by 15%.  
9. Find the new highest quarterly sales.  
10. Check whether the new highest sales exceed ₹30,000.

Rows = Products A, B, C 
Columns = Q1, Q2, Q3, Q4 
"""

sales = [ 
    [15000, 18000, 21000, 24000], 
    [17000, 19000, 23000, 26000], 
    [14000, 20000, 25000, 28000] 
] 

arr = np.array(sales)

new = arr + (arr * 15) / 100

print(f"Shape : {arr.shape}")
print(f"Ndim : {arr.ndim}")
print(f"Dtype : {arr.dtype}")
print(f"annual sales of each product : {arr.sum(axis=1)}")
print(f"highest annual sales : {arr.sum(axis=1)}")
print(f"Q4 sales of Product B : {arr[1][-1]}")
print(f"Q1 and Q2 sales of all products : {arr[:,:2]}")
print(f"highest quarterly sales : {arr.max()}")
print(f"Increased : {new}")
print(f"new highest quarterly sales : {new.max()}")

if new.max() > 30000:
    print("Yes, the new highest sales exceed ₹30,000")
else:
    print("No, the new highest sales doesn't exceed ₹30,000")
```
