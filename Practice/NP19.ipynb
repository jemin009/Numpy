```python
import numpy as np
import pandas as pd

df = pd.read_csv(r"D:\Practice\Numpy\Practice\Dataset\4train.csv")

df
```

```python
#1 
"""
Using Pandas, calculate the following for each Pclass: 
 Average Fare  
 Median Fare  
 Minimum Fare  
 Maximum Fare  
 Average Age 
"""

grp = df.groupby('Pclass').agg(
                                AvgFare = ('Fare','mean'),
                                MedianFare = ('Fare','median'),
                                MinFare = ('Fare','min'),
                                MaxFare = ('Fare','max'),
                                AvgAge = ('Age','mean')
                                )

print(grp)
```

```python
#2
"""
Using Pandas string operations, find all passengers whose Name contains "Mrs.". 
 
Display: 
Name, Age, Fare, Survived
"""

mrs = df.loc[df['Name'].str.contains('Mrs.'),['Name','Age','Fare','Survived']]

print(mrs)
```

```python
#3
"""
Create a new column: 
Pclass_Avg_Fare 
It should contain the average Fare of that passenger's Pclass for every passenger. 
 
Then display: 
Name, Pclass, Fare, Pclass_Avg_Fare 
for the first 10 passengers.
"""

df['Pclass_Avg_Fare'] = df.groupby('Pclass')['Fare'].transform('mean')

show = df.loc[df['Pclass_Avg_Fare'].head(10).index, ['Name','Pclass','Fare','Pclass_Avg_Fare']]

print(show)

```

```python
#4 
"""
Create a pivot table showing the average Fare for each combination of: 
 Rows → Pclass  
 Columns → Sex 
"""

pt = df.pivot_table(
    values='Fare',
    index='Pclass',
    columns='Sex',
    aggfunc='mean'
)

print(pt)
```

```python
#5
"""
Create: 
AdjustedFare 
 
Rules: 
 Fare > 100 → increase by 15%  
 Fare ≤ 100 → increase by 5%  
Then display the 10 passengers with the highest AdjustedFare.
"""

df['AdjustedFare'] = np.where(
    df['Fare'] > 100,
    df['Fare'] * 1.15,
    df['Fare'] * 1.05
)

sort = df.sort_values('AdjustedFare', ascending=False)[-10:][::-1]

print(sort)
```

```python
#6 
"""
Convert Age to a NumPy array. 
Replace missing Age values with: 
median Age of the entire dataset 
 
Then create a new DataFrame column: 
Age_Filled 
 
Display: 
Name, Age, Age_Filled 
for 10 passengers who originally had missing Age.
"""

age = df['Age'].to_numpy()

df['Age_Filled'] = np.where(np.isnan(age), np.nanmedian(age), age)

show = df.loc[
        df['Age'].isna(),
        ['Name', 'Age', 'Age_Filled']
        ].head(10)

print(show)
```

```python
#7 
"""
Create a separate DataFrame containing: 
Pclass 
Average_Fare 
 
where Average_Fare is the average Fare for each Pclass. 
Then merge this DataFrame back into the original Titanic DataFrame using Pclass. 
 
Display: 
Name, Pclass, Fare, Average_Fare 
for the first 10 passengers.
"""

sep = df.groupby('Pclass', as_index=False)['Fare'].mean()

sep = sep.rename(columns={'Fare': 'Average_Fare'})

join = df.merge(sep, on='Pclass')

print(join[['Name', 'Pclass', 'Fare', 'Average_Fare']].head(10))
```

```python
#8
"""
Create a NumPy matrix containing: 
Pclass 
Age 
SibSp 
Parch 
Fare 
 
Requirements: 
 Replace missing Age with median Age.  
 Create the matrix using NumPy.  
 Print its shape.  
 Display the first 5 rows. 
"""

pclass = df['Pclass'].to_numpy()
age = df['Age'].to_numpy()
sibsp = df['SibSp'].to_numpy()
parch = df['Parch'].to_numpy()
fare = df['Fare'].to_numpy()

age = np.where(np.isnan(age), np.nanmedian(age), age)

mat = np.column_stack((pclass, age, sibsp, parch, fare))

print(mat.shape)

print(mat[:5])
```

```python
#9 
"""
Create: 
Fare_Percent_of_Class_Avg 
 
Formula: 
(Fare / Pclass average Fare) × 100 
 
Use transform() to calculate the Pclass average Fare. 
 
Then display the 5 passengers with the highest percentage: 
Name, Pclass, Fare, Fare_Percent_of_Class_Avg
"""

Pclass_avg_Fare = df.groupby('Pclass')['Fare'].transform('mean')

df['Fare_Percent_of_Class_Avg'] = (df['Fare'] / Pclass_avg_Fare) * 100

sort = df.sort_values('Fare_Percent_of_Class_Avg', ascending=False).head(5)

show = sort[['Name','Pclass','Fare','Fare_Percent_of_Class_Avg']]

print(show)
```

```python
#10
"""
Create a final feature DataFrame containing: 
Survived 
Pclass 
Age 
SibSp 
Parch 
Fare 
FamilySize 
FarePerPerson 
PclassAvgFare 
FarePercentOfClassAvg
"""

df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['FarePerPerson'] = df['Fare'] / df['FamilySize']
df['PclassAvgFare'] = df.groupby('Pclass')['Fare'].transform('mean')
df['FarePercentOfClassAvg'] = (df['Fare'] / df['PclassAvgFare']) * 100

show = df[['Survived','Pclass','Age','SibSp','Parch','Fare','FamilySize','FarePerPerson','Pclass_avg_Fare','FarePercentOfClassAvg']]

print(show)
```
